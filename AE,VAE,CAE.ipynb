```python

# CS3807 - Deep Learning Laboratory
# Experiment 7: Autoencoders, Convolutional Autoencoders,
# Denoising Autoencoders and Variational Autoencoders
# Covers Tasks/Sections 1-26 from the given laboratory PDF.

import os
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras import layers, Model
from tensorflow.keras.datasets import mnist
from tensorflow.keras import backend as K
from skimage.metrics import structural_similarity as ssim

# ============================================================
# 1. SETUP
# ============================================================

SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)

TRAIN_SIZE = 10000
TEST_SIZE = 2000
BATCH_SIZE = 128
EPOCHS = 20

# Load MNIST
(x_train_full, y_train_full), (x_test_full, y_test_full) = mnist.load_data()

# Recommended subset from the experiment
x_train = x_train_full[:TRAIN_SIZE].astype("float32") / 255.0
y_train = y_train_full[:TRAIN_SIZE]

x_test = x_test_full[:TEST_SIZE].astype("float32") / 255.0
y_test = y_test_full[:TEST_SIZE]

# Add channel dimension for convolutional models
x_train_cnn = np.expand_dims(x_train, axis=-1)
x_test_cnn = np.expand_dims(x_test, axis=-1)

print("Training shape:", x_train_cnn.shape)
print("Test shape:", x_test_cnn.shape)


# ============================================================
# 2. COMMON METRICS
# ============================================================

def calculate_metrics(original, reconstructed):
    """
    Calculates test-set MSE, MAE and mean SSIM.
    Images are expected in [0, 1].
    """
    original = np.asarray(original)
    reconstructed = np.asarray(reconstructed)

    mse = np.mean((original - reconstructed) ** 2)
    mae = np.mean(np.abs(original - reconstructed))

    # SSIM image-by-image
    ssim_values = []

    for i in range(len(original)):
        img1 = np.squeeze(original[i])
        img2 = np.squeeze(reconstructed[i])

        ssim_values.append(
            ssim(img1, img2, data_range=1.0)
        )

    mean_ssim = np.mean(ssim_values)

    return mse, mae, mean_ssim


def print_metrics(name, mse, mae, mean_ssim):
    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)
    print(f"MSE  : {mse:.6f}")
    print(f"MAE  : {mae:.6f}")
    print(f"SSIM : {mean_ssim:.6f}")


# ============================================================
# 3. PLOT 1 - ORIGINAL VS RECONSTRUCTED
# ============================================================

def plot_original_reconstruction(original, reconstructed, title, n=10):
    n = min(n, len(original))

    plt.figure(figsize=(16, 4))

    for i in range(n):
        plt.subplot(2, n, i + 1)
        plt.imshow(np.squeeze(original[i]), cmap="gray")
        plt.axis("off")

        if i == 0:
            plt.ylabel("Original")

        plt.subplot(2, n, n + i + 1)
        plt.imshow(np.squeeze(reconstructed[i]), cmap="gray")
        plt.axis("off")

        if i == 0:
            plt.ylabel("Reconstructed")

    plt.suptitle(title)
    plt.tight_layout()
    plt.show()


# ============================================================
# 4. FULLY CONNECTED AUTOENCODER
# ============================================================

def build_fc_autoencoder(latent_dim=16):
    input_layer = layers.Input(shape=(784,))

    x = layers.Dense(128, activation="relu")(input_layer)
    x = layers.Dense(32, activation="relu")(x)

    latent = layers.Dense(latent_dim, activation="relu", name="latent")(x)

    x = layers.Dense(32, activation="relu")(latent)
    x = layers.Dense(128, activation="relu")(x)

    output_layer = layers.Dense(784, activation="sigmoid")(x)

    model = Model(input_layer, output_layer, name=f"FC_AE_{latent_dim}")
    encoder = Model(input_layer, latent, name=f"FC_Encoder_{latent_dim}")

    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
        loss="binary_crossentropy"
    )

    return model, encoder


print("\nTraining Fully Connected Autoencoder...")

fc_start = time.time()

fc_ae, fc_encoder = build_fc_autoencoder(latent_dim=16)

fc_history = fc_ae.fit(
    x_train.reshape(-1, 784),
    x_train.reshape(-1, 784),
    validation_data=(
        x_test.reshape(-1, 784),
        x_test.reshape(-1, 784)
    ),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    shuffle=True,
    verbose=1
)

fc_time = time.time() - fc_start

fc_reconstructed = fc_ae.predict(
    x_test.reshape(-1, 784),
    batch_size=BATCH_SIZE,
    verbose=0
)

fc_reconstructed = fc_reconstructed.reshape(-1, 28, 28, 1)

fc_mse, fc_mae, fc_ssim = calculate_metrics(
    x_test_cnn,
    fc_reconstructed
)

print_metrics("Fully Connected Autoencoder", fc_mse, fc_mae, fc_ssim)

print("FC Autoencoder parameters:", fc_ae.count_params())
print(f"FC training time: {fc_time:.2f} seconds")

# Plot 1
plot_original_reconstruction(
    x_test_cnn,
    fc_reconstructed,
    "Plot 1: Original vs Fully Connected AE Reconstruction"
)


# ============================================================
# 5. PLOT 2 - FC TRAINING / VALIDATION LOSS
# ============================================================

plt.figure(figsize=(8, 5))
plt.plot(fc_history.history["loss"], label="Training Loss")
plt.plot(fc_history.history["val_loss"], label="Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Reconstruction Loss")
plt.title("Plot 2: FC Autoencoder Training and Validation Loss")
plt.legend()
plt.grid(True)
plt.show()


# ============================================================
# 6. CONVOLUTIONAL AUTOENCODER
# ============================================================

def build_conv_autoencoder():
    input_layer = layers.Input(shape=(28, 28, 1))

    # Encoder
    x = layers.Conv2D(
        32, 3, activation="relu", padding="same"
    )(input_layer)

    x = layers.MaxPooling2D(
        2, padding="same"
    )(x)

    x = layers.Conv2D(
        64, 3, activation="relu", padding="same"
    )(x)

    x = layers.MaxPooling2D(
        2, padding="same"
    )(x)

    # Latent feature map
    latent = layers.Conv2D(
        64, 3, activation="relu", padding="same",
        name="latent_feature_map"
    )(x)

    # Decoder
    x = layers.UpSampling2D(2)(latent)

    x = layers.Conv2D(
        32, 3, activation="relu", padding="same"
    )(x)

    x = layers.UpSampling2D(2)(x)

    output_layer = layers.Conv2D(
        1, 3, activation="sigmoid", padding="same"
    )(x)

    model = Model(
        input_layer,
        output_layer,
        name="Convolutional_Autoencoder"
    )

    model.compile(
        optimizer="adam",
        loss="binary_crossentropy"
    )

    return model


print("\nTraining Convolutional Autoencoder...")

cae_start = time.time()

cae = build_conv_autoencoder()

cae_history = cae.fit(
    x_train_cnn,
    x_train_cnn,
    validation_data=(x_test_cnn, x_test_cnn),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    shuffle=True,
    verbose=1
)

cae_time = time.time() - cae_start

cae_reconstructed = cae.predict(
    x_test_cnn,
    batch_size=BATCH_SIZE,
    verbose=0
)

cae_mse, cae_mae, cae_ssim = calculate_metrics(
    x_test_cnn,
    cae_reconstructed
)

print_metrics("Convolutional Autoencoder", cae_mse, cae_mae, cae_ssim)

print("CAE parameters:", cae.count_params())
print(f"CAE training time: {cae_time:.2f} seconds")


# ============================================================
# 7. PLOT 3 - FC AE VS CAE
# ============================================================

n = 10

plt.figure(figsize=(16, 6))

for i in range(n):
    plt.subplot(3, n, i + 1)
    plt.imshow(x_test[i], cmap="gray")
    plt.axis("off")

    if i == 0:
        plt.ylabel("Original")

    plt.subplot(3, n, n + i + 1)
    plt.imshow(np.squeeze(fc_reconstructed[i]), cmap="gray")
    plt.axis("off")

    if i == 0:
        plt.ylabel("FC-AE")

    plt.subplot(3, n, 2 * n + i + 1)
    plt.imshow(np.squeeze(cae_reconstructed[i]), cmap="gray")
    plt.axis("off")

    if i == 0:
        plt.ylabel("CAE")

plt.suptitle(
    "Plot 3: Original -> Fully Connected AE -> Convolutional AE"
)
plt.tight_layout()
plt.show()


# ============================================================
# 8. DENOISING AUTOENCODER
# ============================================================

def add_gaussian_noise(images, sigma=0.2):
    noise = np.random.normal(
        loc=0.0,
        scale=sigma,
        size=images.shape
    )

    noisy = images + noise
    noisy = np.clip(noisy, 0.0, 1.0)

    return noisy.astype("float32")


def add_salt_pepper_noise(images, p=0.1):
    noisy = images.copy()

    random_values = np.random.random(images.shape)

    noisy[random_values < (p / 2)] = 0.0
    noisy[random_values > 1 - (p / 2)] = 1.0

    return noisy.astype("float32")


def build_denoising_cae():
    input_layer = layers.Input(shape=(28, 28, 1))

    x = layers.Conv2D(
        32, 3, activation="relu", padding="same"
    )(input_layer)

    x = layers.MaxPooling2D(
        2, padding="same"
    )(x)

    x = layers.Conv2D(
        64, 3, activation="relu", padding="same"
    )(x)

    x = layers.MaxPooling2D(
        2, padding="same"
    )(x)

    x = layers.Conv2D(
        64, 3, activation="relu", padding="same"
    )(x)

    x = layers.UpSampling2D(2)(x)

    x = layers.Conv2D(
        32, 3, activation="relu", padding="same"
    )(x)

    x = layers.UpSampling2D(2)(x)

    output_layer = layers.Conv2D(
        1, 3, activation="sigmoid", padding="same"
    )(x)

    model = Model(
        input_layer,
        output_layer,
        name="Denoising_CAE"
    )

    model.compile(
        optimizer="adam",
        loss="binary_crossentropy"
    )

    return model


print("\nCreating noisy training data...")

# Main denoising experiment uses Gaussian sigma = 0.2.
# The PDF also asks for sigma 0.1, 0.2 and 0.3 for evaluation.
noisy_train = add_gaussian_noise(
    x_train_cnn,
    sigma=0.2
)

print("\nTraining Denoising Convolutional Autoencoder...")

denoise_start = time.time()

denoise_cae = build_denoising_cae()

denoise_history = denoise_cae.fit(
    noisy_train,
    x_train_cnn,            # clean target
    validation_data=(x_test_cnn, x_test_cnn),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    shuffle=True,
    verbose=1
)

denoise_time = time.time() - denoise_start

# Evaluate on sigma=0.2
noisy_test_02 = add_gaussian_noise(
    x_test_cnn,
    sigma=0.2
)

denoised_test = denoise_cae.predict(
    noisy_test_02,
    batch_size=BATCH_SIZE,
    verbose=0
)

denoise_mse, denoise_mae, denoise_ssim = calculate_metrics(
    x_test_cnn,
    denoised_test
)

print_metrics(
    "Denoising CAE (Gaussian sigma=0.2)",
    denoise_mse,
    denoise_mae,
    denoise_ssim
)

print("Denoising CAE parameters:", denoise_cae.count_params())
print(f"Denoising training time: {denoise_time:.2f} seconds")


# ============================================================
# 9. PLOT 4 - CLEAN VS NOISY VS DENOISED
# ============================================================

plt.figure(figsize=(16, 6))

for i in range(n):
    plt.subplot(3, n, i + 1)
    plt.imshow(np.squeeze(x_test_cnn[i]), cmap="gray")
    plt.axis("off")

    if i == 0:
        plt.ylabel("Clean")

    plt.subplot(3, n, n + i + 1)
    plt.imshow(np.squeeze(noisy_test_02[i]), cmap="gray")
    plt.axis("off")

    if i == 0:
        plt.ylabel("Noisy")

    plt.subplot(3, n, 2 * n + i + 1)
    plt.imshow(np.squeeze(denoised_test[i]), cmap="gray")
    plt.axis("off")

    if i == 0:
        plt.ylabel("Denoised")

plt.suptitle(
    "Plot 4: Clean vs Noisy vs Denoised"
)
plt.tight_layout()
plt.show()


# ============================================================
# 10. PLOT 5 - GAUSSIAN NOISE LEVEL STUDY
# ============================================================

noise_levels = [0.1, 0.2, 0.3]
noise_results = []

print("\nGaussian noise-level evaluation:")

for sigma in noise_levels:
    noisy_test = add_gaussian_noise(
        x_test_cnn,
        sigma=sigma
    )

    denoised = denoise_cae.predict(
        noisy_test,
        batch_size=BATCH_SIZE,
        verbose=0
    )

    mse, mae, mean_ssim = calculate_metrics(
        x_test_cnn,
        denoised
    )

    noise_results.append(
        [sigma, mse, mae, mean_ssim]
    )

    print(
        f"sigma={sigma:.1f} | "
        f"MSE={mse:.6f} | "
        f"MAE={mae:.6f} | "
        f"SSIM={mean_ssim:.6f}"
    )

noise_df = pd.DataFrame(
    noise_results,
    columns=["Noise Level", "MSE", "MAE", "SSIM"]
)

print("\nNoise Level Results")
print(noise_df.to_string(index=False))

# Separate figures because the experiment says not to mix incompatible scales.
plt.figure(figsize=(8, 5))
plt.plot(noise_df["Noise Level"], noise_df["MSE"], marker="o")
plt.xlabel("Gaussian Noise Sigma")
plt.ylabel("MSE")
plt.title("Plot 5A: Noise Level vs MSE")
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(noise_df["Noise Level"], noise_df["MAE"], marker="o")
plt.xlabel("Gaussian Noise Sigma")
plt.ylabel("MAE")
plt.title("Plot 5B: Noise Level vs MAE")
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(noise_df["Noise Level"], noise_df["SSIM"], marker="o")
plt.xlabel("Gaussian Noise Sigma")
plt.ylabel("SSIM")
plt.title("Plot 5C: Noise Level vs SSIM")
plt.grid(True)
plt.show()


# ============================================================
# 11. SALT-AND-PEPPER NOISE (REQUIRED BY TASK 13)
# ============================================================

sp_levels = [0.05, 0.10, 0.20]
sp_results = []

for p in sp_levels:
    noisy_sp = add_salt_pepper_noise(
        x_test_cnn,
        p=p
    )

    denoised_sp = denoise_cae.predict(
        noisy_sp,
        batch_size=BATCH_SIZE,
        verbose=0
    )

    mse, mae, mean_ssim = calculate_metrics(
        x_test_cnn,
        denoised_sp
    )

    sp_results.append(
        [p, mse, mae, mean_ssim]
    )

sp_df = pd.DataFrame(
    sp_results,
    columns=["Salt-Pepper Level", "MSE", "MAE", "SSIM"]
)

print("\nSalt-and-Pepper Noise Results")
print(sp_df.to_string(index=False))


# ============================================================
# 12. VARIATIONAL AUTOENCODER
# ============================================================

LATENT_DIM = 2


class Sampling(layers.Layer):
    """
    Reparameterization trick:
    z = mu + sigma * epsilon
    epsilon ~ N(0, I)
    """

    def call(self, inputs):
        z_mean, z_log_var = inputs

        epsilon = tf.random.normal(
            shape=tf.shape(z_mean)
        )

        return z_mean + tf.exp(
            0.5 * z_log_var
        ) * epsilon


def build_vae(latent_dim=2):
    # ---------------- Encoder ----------------
    encoder_inputs = layers.Input(
        shape=(28, 28, 1),
        name="encoder_input"
    )

    x = layers.Conv2D(
        32, 3, activation="relu", strides=2,
        padding="same"
    )(encoder_inputs)

    x = layers.Conv2D(
        64, 3, activation="relu", strides=2,
        padding="same"
    )(x)

    x = layers.Flatten()(x)

    x = layers.Dense(128, activation="relu")(x)

    z_mean = layers.Dense(
        latent_dim,
        name="z_mean"
    )(x)

    z_log_var = layers.Dense(
        latent_dim,
        name="z_log_var"
    )(x)

    z = Sampling()([z_mean, z_log_var])

    encoder = Model(
        encoder_inputs,
        [z_mean, z_log_var, z],
        name="VAE_Encoder"
    )

    # ---------------- Decoder ----------------
    latent_inputs = layers.Input(
        shape=(latent_dim,),
        name="z_sampling"
    )

    x = layers.Dense(
        7 * 7 * 64,
        activation="relu"
    )(latent_inputs)

    x = layers.Reshape((7, 7, 64))(x)

    x = layers.Conv2DTranspose(
        64, 3, activation="relu",
        strides=2, padding="same"
    )(x)

    x = layers.Conv2DTranspose(
        32, 3, activation="relu",
        strides=2, padding="same"
    )(x)

    decoder_outputs = layers.Conv2DTranspose(
        1, 3, activation="sigmoid",
        padding="same"
    )(x)

    decoder = Model(
        latent_inputs,
        decoder_outputs,
        name="VAE_Decoder"
    )

    return encoder, decoder


class VAE(Model):

    def __init__(self, encoder, decoder, **kwargs):
        super().__init__(**kwargs)

        self.encoder = encoder
        self.decoder = decoder

        self.total_loss_tracker = tf.keras.metrics.Mean(
            name="total_loss"
        )

        self.reconstruction_loss_tracker = tf.keras.metrics.Mean(
            name="reconstruction_loss"
        )

        self.kl_loss_tracker = tf.keras.metrics.Mean(
            name="kl_loss"
        )

    @property
    def metrics(self):
        return [
            self.total_loss_tracker,
            self.reconstruction_loss_tracker,
            self.kl_loss_tracker
        ]

    def train_step(self, data):

        if isinstance(data, tuple):
            data = data[0]

        with tf.GradientTape() as tape:

            z_mean, z_log_var, z = self.encoder(
                data,
                training=True
            )

            reconstruction = self.decoder(
                z,
                training=True
            )

            # Reconstruction loss
            reconstruction_loss = tf.reduce_mean(
                tf.reduce_sum(
                    tf.keras.losses.binary_crossentropy(
                        data,
                        reconstruction
                    ),
                    axis=(1, 2)
                )
            )

            # KL divergence
            kl_loss = -0.5 * tf.reduce_mean(
                tf.reduce_sum(
                    1
                    + z_log_var
                    - tf.square(z_mean)
                    - tf.exp(z_log_var),
                    axis=1
                )
            )

            total_loss = (
                reconstruction_loss
                + kl_loss
            )

        grads = tape.gradient(
            total_loss,
            self.trainable_weights
        )

        self.optimizer.apply_gradients(
            zip(grads, self.trainable_weights)
        )

        self.total_loss_tracker.update_state(
            total_loss
        )

        self.reconstruction_loss_tracker.update_state(
            reconstruction_loss
        )

        self.kl_loss_tracker.update_state(
            kl_loss
        )

        return {
            "loss": self.total_loss_tracker.result(),
            "reconstruction_loss":
                self.reconstruction_loss_tracker.result(),
            "kl_loss":
                self.kl_loss_tracker.result()
        }

    def test_step(self, data):

        if isinstance(data, tuple):
            data = data[0]

        z_mean, z_log_var, z = self.encoder(
            data,
            training=False
        )

        reconstruction = self.decoder(
            z,
            training=False
        )

        reconstruction_loss = tf.reduce_mean(
            tf.reduce_sum(
                tf.keras.losses.binary_crossentropy(
                    data,
                    reconstruction
                ),
                axis=(1, 2)
            )
        )

        kl_loss = -0.5 * tf.reduce_mean(
            tf.reduce_sum(
                1
                + z_log_var
                - tf.square(z_mean)
                - tf.exp(z_log_var),
                axis=1
            )
        )

        total_loss = (
            reconstruction_loss
            + kl_loss
        )

        self.total_loss_tracker.update_state(
            total_loss
        )

        self.reconstruction_loss_tracker.update_state(
            reconstruction_loss
        )

        self.kl_loss_tracker.update_state(
            kl_loss
        )

        return {
            "loss": self.total_loss_tracker.result(),
            "reconstruction_loss":
                self.reconstruction_loss_tracker.result(),
            "kl_loss":
                self.kl_loss_tracker.result()
        }


print("\nBuilding VAE...")

vae_encoder, vae_decoder = build_vae(
    latent_dim=LATENT_DIM
)

vae = VAE(
    vae_encoder,
    vae_decoder,
    name="VAE"
)

vae.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-3
    )
)

print("\nTraining VAE...")

vae_start = time.time()

vae_history = vae.fit(
    x_train_cnn,
    validation_data=(x_test_cnn,),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    shuffle=True,
    verbose=1
)

vae_time = time.time() - vae_start


# ============================================================
# 13. VAE TEST RECONSTRUCTION + LOSSES
# ============================================================

z_mean_test, z_log_var_test, z_test = vae_encoder.predict(
    x_test_cnn,
    batch_size=BATCH_SIZE,
    verbose=0
)

vae_reconstructed = vae_decoder.predict(
    z_test,
    batch_size=BATCH_SIZE,
    verbose=0
)

vae_mse, vae_mae, vae_ssim = calculate_metrics(
    x_test_cnn,
    vae_reconstructed
)

# Evaluate using deterministic mean latent vector for a stable
# reconstruction-loss measurement.
vae_mean_reconstruction = vae_decoder.predict(
    z_mean_test,
    batch_size=BATCH_SIZE,
    verbose=0
)

binary_crossentropy_values = tf.keras.losses.binary_crossentropy(
    x_test_cnn,
    vae_mean_reconstruction
)

vae_reconstruction_loss = float(
    tf.reduce_mean(
        tf.reduce_sum(
            binary_crossentropy_values,
            axis=(1, 2)
        )
    ).numpy()
)

vae_kl_loss = float(
    -0.5 * np.mean(
        np.sum(
            1
            + z_log_var_test
            - np.square(z_mean_test)
            - np.exp(z_log_var_test),
            axis=1
        )
    )
)

vae_total_loss = (
    vae_reconstruction_loss
    + vae_kl_loss
)

print("\n" + "=" * 60)
print("VAE RESULTS")
print("=" * 60)

print(f"Reconstruction Loss: {vae_reconstruction_loss:.6f}")
print(f"KL Loss            : {vae_kl_loss:.6f}")
print(f"Total Loss         : {vae_total_loss:.6f}")
print(f"Test MSE           : {vae_mse:.6f}")
print(f"Test MAE           : {vae_mae:.6f}")
print(f"Mean SSIM          : {vae_ssim:.6f}")

vae_parameters = (
    vae_encoder.count_params()
    + vae_decoder.count_params()
)

print("VAE parameters      :", vae_parameters)
print(f"VAE training time   : {vae_time:.2f} seconds")


# ============================================================
# 14. PLOT 6 - VAE LATENT SPACE
# ============================================================

plt.figure(figsize=(8, 7))

scatter = plt.scatter(
    z_mean_test[:, 0],
    z_mean_test[:, 1],
    c=y_test,
    cmap="tab10",
    s=8,
    alpha=0.7
)

plt.colorbar(scatter, ticks=range(10))
plt.xlabel("z1")
plt.ylabel("z2")
plt.title("Plot 6: VAE Latent Space")
plt.grid(True)
plt.show()


# ============================================================
# 15. PLOT 7 - RANDOMLY GENERATED IMAGES
# ============================================================

num_generated = 25

random_latent = np.random.normal(
    size=(num_generated, LATENT_DIM)
).astype("float32")

generated_images = vae_decoder.predict(
    random_latent,
    verbose=0
)

plt.figure(figsize=(8, 8))

for i in range(num_generated):
    plt.subplot(5, 5, i + 1)
    plt.imshow(
        np.squeeze(generated_images[i]),
        cmap="gray"
    )
    plt.axis("off")

plt.suptitle(
    "Plot 7: Randomly Generated VAE Images"
)
plt.tight_layout()
plt.show()


# ============================================================
# 16. PLOT 8 - LATENT SPACE INTERPOLATION
# ============================================================

# Select two test latent vectors
zA = z_mean_test[0]
zB = z_mean_test[1]

alphas = np.linspace(
    0.0,
    1.0,
    11
)

interpolated_z = np.array([
    (1 - alpha) * zA + alpha * zB
    for alpha in alphas
]).astype("float32")

interpolated_images = vae_decoder.predict(
    interpolated_z,
    verbose=0
)

plt.figure(figsize=(18, 3))

for i in range(len(alphas)):
    plt.subplot(1, len(alphas), i + 1)
    plt.imshow(
        np.squeeze(interpolated_images[i]),
        cmap="gray"
    )
    plt.title(f"{alphas[i]:.1f}")
    plt.axis("off")

plt.suptitle(
    "Plot 8: VAE Latent-Space Interpolation"
)
plt.tight_layout()
plt.show()


# ============================================================
# 17. PLOT 9 - VAE TRAINING / VALIDATION RECONSTRUCTION LOSS
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    vae_history.history["reconstruction_loss"],
    label="Training Reconstruction Loss"
)

if "val_reconstruction_loss" in vae_history.history:
    plt.plot(
        vae_history.history["val_reconstruction_loss"],
        label="Validation Reconstruction Loss"
    )

plt.xlabel("Epoch")
plt.ylabel("Reconstruction Loss")
plt.title(
    "Plot 9: VAE Training and Validation Reconstruction Loss"
)
plt.legend()
plt.grid(True)
plt.show()


# ============================================================
# 18. RECONSTRUCTION ERROR DISTRIBUTION
# ============================================================

# Use CAE for the reconstruction-error analysis.
per_image_errors = np.mean(
    (
        x_test_cnn - cae_reconstructed
    ) ** 2,
    axis=(1, 2, 3)
)

# Plot 9 in the experiment's section 23
plt.figure(figsize=(8, 5))

plt.hist(
    per_image_errors,
    bins=40
)

plt.xlabel("Per-image Reconstruction MSE")
plt.ylabel("Number of Images")
plt.title(
    "Reconstruction Error Distribution on Test Set"
)
plt.grid(True)
plt.show()


# ============================================================
# 19. FIVE HIGHEST-ERROR IMAGES
# ============================================================

highest_error_indices = np.argsort(
    per_image_errors
)[-5:][::-1]

print("\nFive highest reconstruction-error samples:")

for rank, idx in enumerate(
    highest_error_indices,
    start=1
):
    print(
        f"Rank {rank}: "
        f"Test index={idx}, "
        f"Digit={y_test[idx]}, "
        f"Error={per_image_errors[idx]:.6f}"
    )


plt.figure(figsize=(12, 5))

for i, idx in enumerate(
    highest_error_indices
):

    plt.subplot(2, 5, i + 1)

    plt.imshow(
        np.squeeze(x_test_cnn[idx]),
        cmap="gray"
    )

    plt.title(
        f"Original\nDigit={y_test[idx]}"
    )

    plt.axis("off")

    plt.subplot(
        2, 5,
        5 + i + 1
    )

    plt.imshow(
        np.squeeze(cae_reconstructed[idx]),
        cmap="gray"
    )

    plt.title(
        f"Reconstruction\nErr={per_image_errors[idx]:.4f}"
    )

    plt.axis("off")

plt.suptitle(
    "Five Highest Reconstruction-Error Samples"
)
plt.tight_layout()
plt.show()


# ============================================================
# 20. CONSOLIDATED RESULTS - TASK 25
# ============================================================

# ============================================================
# TASK 25 - CONSOLIDATED RESULTS
# ============================================================

# Calculate parameter counts from the already-trained models
fc_parameters = fc_ae.count_params()
cae_parameters = cae.count_params()
denoise_parameters = denoise_cae.count_params()

# VAE is a custom wrapper model, so count encoder + decoder parameters
vae_parameters = (
    vae_encoder.count_params()
    + vae_decoder.count_params()
)

results = {
    "Model": [
        "FC Autoencoder",
        "Convolutional Autoencoder",
        "Denoising CAE",
        "VAE"
    ],

    "MSE": [
        fc_mse,
        cae_mse,
        denoise_mse,
        vae_mse
    ],

    "MAE": [
        fc_mae,
        cae_mae,
        denoise_mae,
        vae_mae
    ],

    "SSIM": [
        fc_ssim,
        cae_ssim,
        denoise_ssim,
        vae_ssim
    ],

    "Parameters": [
        fc_parameters,
        cae_parameters,
        denoise_parameters,
        vae_parameters
    ],

    "Time (sec)": [
        fc_time,
        cae_time,
        denoise_time,
        vae_time
    ]
}

results_df = pd.DataFrame(results)

print("\n" + "=" * 80)
print("TASK 25 - CONSOLIDATED RESULTS")
print("=" * 80)

display(results_df)

# ============================================================
# 21. VAE LOSS TABLE
# ============================================================

vae_loss_df = pd.DataFrame([
    {
        "VAE Metric": "Reconstruction Loss",
        "Value": vae_reconstruction_loss
    },
    {
        "VAE Metric": "KL Loss",
        "Value": vae_kl_loss
    },
    {
        "VAE Metric": "Total Loss",
        "Value": vae_total_loss
    },
    {
        "VAE Metric": "Test MSE",
        "Value": vae_mse
    },
    {
        "VAE Metric": "Test MAE",
        "Value": vae_mae
    },
    {
        "VAE Metric": "Mean SSIM",
        "Value": vae_ssim
    }
])

print("\nVAE Loss / Metric Table")
print(vae_loss_df.to_string(index=False))


# ============================================================
# 22. TASK 26 - REQUIRED INFERENCES
# ============================================================

print("\n" + "=" * 80)
print("TASK 26 - REQUIRED INFERENCES")
print("=" * 80)

print("""
1. EFFECT OF LATENT DIMENSION
The latent vector controls the amount of information that can pass from
the encoder to the decoder. A smaller latent dimension produces stronger
compression and may lose fine details. Increasing the latent dimension
generally gives the model more capacity to preserve image information.

2. FULLY CONNECTED VS CONVOLUTIONAL RECONSTRUCTION
The fully connected autoencoder first flattens the image into 784 values,
so the original spatial arrangement is not explicitly preserved by the
network structure. The convolutional autoencoder operates directly on
the 2-D image and uses local receptive fields, allowing it to exploit
spatial patterns such as edges and shapes.

3. EFFECT OF GAUSSIAN NOISE LEVEL
As Gaussian noise increases, the input becomes increasingly corrupted.
The reconstruction metrics should be inspected to determine how MSE,
MAE and SSIM change. In general, stronger corruption makes the denoising
task more difficult because more information has to be recovered.

4. ABILITY OF THE DENOISING AUTOENCODER TO RECOVER IMAGE STRUCTURE
The denoising autoencoder is trained with noisy images as input but clean
images as targets. Therefore, it learns to remove corruption while
preserving important digit structure. The clean/noisy/denoised plot should
be used to comment on how well major strokes and shapes are recovered.

5. CHARACTERISTICS OF THE VAE LATENT DISTRIBUTION
Unlike a conventional autoencoder, the VAE encoder produces a mean and
log-variance and samples a latent vector from the resulting distribution.
The KL-divergence term encourages the learned distribution to remain close
to the standard normal prior.

6. STRUCTURE OF THE TWO-DIMENSIONAL VAE LATENT SPACE
The latent-space scatter plot shows where different test digits are mapped
in the two-dimensional latent space. Similar digits may occupy nearby
regions, while different digits may form partially separated or overlapping
regions.

7. QUALITY AND DIVERSITY OF GENERATED SAMPLES
The VAE can generate new images by sampling z from a standard normal
distribution and passing those samples through the decoder. The generated
grid should be inspected for recognizable digit shapes, variation between
samples, and ambiguous or unrealistic outputs.

8. SMOOTHNESS OF LATENT-SPACE INTERPOLATION
Interpolation creates intermediate latent vectors between two endpoints.
If the learned latent space is smooth, the decoded images should change
gradually as alpha moves from 0 to 1 rather than changing abruptly.

9. RECONSTRUCTION ERROR VS VISUAL QUALITY
A higher per-image reconstruction error indicates a larger numerical
difference between the original and reconstructed image. The five
highest-error examples should be visually inspected to determine whether
they contain unusual shapes, ambiguous strokes, thickness variations,
or other patterns that are difficult for the model to reconstruct.

10. DETERMINISTIC VS PROBABILISTIC LATENT REPRESENTATIONS
A conventional autoencoder maps an input to a deterministic latent vector.
A VAE represents an input using a probability distribution described by
its mean and variance. The probabilistic representation makes sampling
and generation possible while the KL term regularizes the latent space.

IMPORTANT:
Replace these general inference statements with observations from your
actual plots and numerical results before submitting the laboratory report.
The numerical values in the report must come from your own execution.
""")


# ============================================================
# 23. OPTIONAL: SAVE RESULTS
# ============================================================

results_df.to_csv(
    "experiment7_consolidated_results.csv",
    index=False
)

noise_df.to_csv(
    "experiment7_gaussian_noise_results.csv",
    index=False
)

sp_df.to_csv(
    "experiment7_salt_pepper_results.csv",
    index=False
)

vae_loss_df.to_csv(
    "experiment7_vae_results.csv",
    index=False
)

print("\nResult CSV files saved in the current folder.")
print("Experiment 7 Tasks 1-26 completed.")
```
