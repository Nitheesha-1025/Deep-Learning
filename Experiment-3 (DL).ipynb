```python
# ============================================================
# CS3807 - Deep Learning Laboratory
# Experiment 3 - CNN Implementation using CIFAR-10
# Tasks 1 to 7 in ONE program
# ============================================================

import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras import layers, models
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# ------------------------------------------------------------
# 0. SETTINGS
# ------------------------------------------------------------

np.random.seed(42)
tf.random.set_seed(42)

os.makedirs("outputs", exist_ok=True)

class_names = [
    "airplane",
    "automobile",
    "bird",
    "cat",
    "deer",
    "dog",
    "frog",
    "horse",
    "ship",
    "truck"
]

print("=" * 70)
print("CS3807 - DEEP LEARNING LABORATORY")
print("EXPERIMENT 3 - CNN ON CIFAR-10")
print("=" * 70)


# ============================================================
# TASK 1
# Load CIFAR-10
# Display 10 sample images
# Print dataset dimensions
# Plot class distribution
# ============================================================

print("\n\n" + "=" * 70)
print("TASK 1 - CIFAR-10 DATASET")
print("=" * 70)

# Load dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

print("\nDataset dimensions:")
print("Training images :", x_train.shape)
print("Training labels :", y_train.shape)
print("Testing images  :", x_test.shape)
print("Testing labels  :", y_test.shape)

print("\nDataset information:")
print("Training samples :", len(x_train))
print("Testing samples  :", len(x_test))
print("Image size       :", x_train.shape[1:])
print("Number of classes:", len(class_names))


# ------------------------------------------------------------
# Display 10 sample images
# ------------------------------------------------------------

plt.figure(figsize=(12, 6))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i])
    plt.title(class_names[int(y_train[i])])
    plt.axis("off")

plt.suptitle("CIFAR-10 Sample Images")
plt.tight_layout()

plt.savefig(
    "outputs/sample_images.jpg",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ------------------------------------------------------------
# Class distribution
# ------------------------------------------------------------

class_counts = np.bincount(
    y_train.flatten(),
    minlength=10
)

print("\nClass distribution:")
for i in range(10):
    print(
        f"{class_names[i]:12s} : "
        f"{class_counts[i]}"
    )

plt.figure(figsize=(10, 6))

plt.bar(class_names, class_counts)

plt.xlabel("Class")
plt.ylabel("Number of Images")
plt.title("CIFAR-10 Class Distribution")
plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    "outputs/class_distribution.jpg",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ============================================================
# NORMALIZATION
# ============================================================

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

print("\nImages normalized to range [0, 1].")


# ============================================================
# TASK 2
# Convolution layer
# Compare 3x3, 5x5 and 7x7
# Record feature map sizes
# ============================================================

print("\n\n" + "=" * 70)
print("TASK 2 - CONVOLUTION KERNEL COMPARISON")
print("=" * 70)

input_shape = (32, 32, 3)

kernel_sizes = [3, 5, 7]

print("\nInput size:", input_shape)

print("\nUsing stride = 1 and padding = valid:")

for k in kernel_sizes:

    conv_model = models.Sequential([
        layers.Input(shape=input_shape),
        layers.Conv2D(
            filters=16,
            kernel_size=(k, k),
            strides=1,
            padding="valid"
        )
    ])

    output_shape = conv_model.output_shape

    print(
        f"Kernel {k}x{k} --> "
        f"Feature map size = "
        f"{output_shape[1]} x {output_shape[2]} x {output_shape[3]}"
    )


# Theoretical calculation
print("\nTheoretical output dimensions:")

for k in kernel_sizes:

    output_size = ((32 - k + 0) // 1) + 1

    print(
        f"{k}x{k}: "
        f"({32} - {k} + 2(0))/1 + 1 "
        f"= {output_size}"
    )


# ------------------------------------------------------------
# Plot kernel comparison
# ------------------------------------------------------------

feature_map_sizes = []

for k in kernel_sizes:
    size = 32 - k + 1
    feature_map_sizes.append(size)

plt.figure(figsize=(8, 5))

plt.bar(
    ["3x3", "5x5", "7x7"],
    feature_map_sizes
)

plt.xlabel("Kernel Size")
plt.ylabel("Feature Map Height/Width")
plt.title("Effect of Kernel Size on Feature Map Size")

plt.tight_layout()

plt.savefig(
    "outputs/kernel_comparison.jpg",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ============================================================
# TASK 3
# Stride = 1 vs Stride = 2
# Padding = Same vs Valid
# ============================================================

print("\n\n" + "=" * 70)
print("TASK 3 - STRIDE AND PADDING")
print("=" * 70)

N = 32
F = 3


def calculate_valid(N, F, S):
    return ((N - F) // S) + 1


print("\nUsing kernel = 3x3")

print("\n1. Stride = 1, Padding = Valid")
print(
    "Output =",
    calculate_valid(32, 3, 1),
    "x",
    calculate_valid(32, 3, 1)
)

print("\n2. Stride = 2, Padding = Valid")
print(
    "Output =",
    calculate_valid(32, 3, 2),
    "x",
    calculate_valid(32, 3, 2)
)

print("\n3. Stride = 1, Padding = Same")

same_s1 = int(np.ceil(32 / 1))

print(
    "Output =",
    same_s1,
    "x",
    same_s1
)

print("\n4. Stride = 2, Padding = Same")

same_s2 = int(np.ceil(32 / 2))

print(
    "Output =",
    same_s2,
    "x",
    same_s2
)


# ------------------------------------------------------------
# Verify using TensorFlow
# ------------------------------------------------------------

print("\nTensorFlow verification:")

settings = [
    ("Stride 1 - Valid", 1, "valid"),
    ("Stride 2 - Valid", 2, "valid"),
    ("Stride 1 - Same", 1, "same"),
    ("Stride 2 - Same", 2, "same")
]

for name, stride, padding in settings:

    model = models.Sequential([
        layers.Input(shape=(32, 32, 3)),
        layers.Conv2D(
            16,
            (3, 3),
            strides=stride,
            padding=padding
        )
    ])

    print(
        f"{name:25s} --> "
        f"{model.output_shape[1]} x "
        f"{model.output_shape[2]} x "
        f"{model.output_shape[3]}"
    )


# ============================================================
# TASK 4
# Feature Map Visualization
# Display at least 8 feature maps
# ============================================================

print("\n\n" + "=" * 70)
print("TASK 4 - FEATURE MAP VISUALIZATION")
print("=" * 70)


# Create first convolution layer
feature_model = models.Sequential([
    layers.Input(shape=(32, 32, 3)),
    layers.Conv2D(
        16,
        (3, 3),
        padding="same",
        activation="relu"
    )
])

# Generate feature maps
sample_image = x_train[0:1]

feature_maps = feature_model.predict(
    sample_image,
    verbose=0
)

print("\nFeature map output shape:")
print(feature_maps.shape)

print(
    "\nDisplaying first 8 feature maps."
)


plt.figure(figsize=(12, 6))

for i in range(8):

    plt.subplot(2, 4, i + 1)

    plt.imshow(
        feature_maps[0, :, :, i],
        cmap="viridis"
    )

    plt.title(f"Feature Map {i + 1}")
    plt.axis("off")

plt.suptitle("Feature Maps After First Convolution")

plt.tight_layout()

plt.savefig(
    "outputs/feature_maps.jpg",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ============================================================
# TASK 5
# Compare Max Pooling and Average Pooling
# ============================================================

print("\n\n" + "=" * 70)
print("TASK 5 - MAX POOLING VS AVERAGE POOLING")
print("=" * 70)


pool_input = x_train[0:1]

max_pool = layers.MaxPooling2D(
    pool_size=(2, 2),
    strides=2
)

avg_pool = layers.AveragePooling2D(
    pool_size=(2, 2),
    strides=2
)

max_output = max_pool(pool_input)
avg_output = avg_pool(pool_input)

print("\nInput shape:")
print(pool_input.shape)

print("\nMax Pooling output:")
print(max_output.shape)

print("\nAverage Pooling output:")
print(avg_output.shape)


# ------------------------------------------------------------
# Visual comparison
# ------------------------------------------------------------

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)

plt.imshow(
    max_output[0]
)

plt.title("Max Pooling")
plt.axis("off")


plt.subplot(1, 2, 2)

plt.imshow(
    avg_output[0]
)

plt.title("Average Pooling")
plt.axis("off")

plt.tight_layout()

plt.savefig(
    "outputs/pooling_comparison.jpg",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ============================================================
# TASK 6
# Construct CNN
#
# Input
# Conv
# ReLU
# MaxPool
# Conv
# ReLU
# MaxPool
# Flatten
# Dense
# Softmax
#
# Optimizer = Adam
# Epochs = 20
# Batch size = 32
# ============================================================

print("\n\n" + "=" * 70)
print("TASK 6 - CNN MODEL TRAINING")
print("=" * 70)


model = models.Sequential([

    # First convolution
    layers.Conv2D(
        16,
        (3, 3),
        padding="same",
        activation="relu",
        input_shape=(32, 32, 3)
    ),

    # First pooling
    layers.MaxPooling2D(
        pool_size=(2, 2)
    ),

    # Second convolution
    layers.Conv2D(
        32,
        (3, 3),
        padding="same",
        activation="relu"
    ),

    # Second pooling
    layers.MaxPooling2D(
        pool_size=(2, 2)
    ),

    # Flatten
    layers.Flatten(),

    # Dense
    layers.Dense(
        128,
        activation="relu"
    ),

    # Output
    layers.Dense(
        10,
        activation="softmax"
    )
])


# ------------------------------------------------------------
# Compile
# ------------------------------------------------------------

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)


print("\nCNN Architecture:")
model.summary()


print(
    "\nTotal trainable parameters:",
    model.count_params()
)


# ------------------------------------------------------------
# Train
# ------------------------------------------------------------

print("\nStarting training...")
print("Epochs    : 20")
print("Batch size: 32")
print("Optimizer : Adam")


history = model.fit(
    x_train,
    y_train,
    epochs=20,
    batch_size=32,
    validation_split=0.1,
    verbose=1
)


# ============================================================
# MANDATORY PLOTS
# Training Accuracy
# Validation Accuracy
# Training Loss
# Validation Loss
# ============================================================

print("\n\n" + "=" * 70)
print("GENERATING TRAINING PLOTS")
print("=" * 70)


# ------------------------------------------------------------
# Accuracy plot
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Training and Validation Accuracy")
plt.legend()
plt.grid(True)

plt.tight_layout()

plt.savefig(
    "outputs/accuracy_plot.jpg",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ------------------------------------------------------------
# Loss plot
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training and Validation Loss")
plt.legend()
plt.grid(True)

plt.tight_layout()

plt.savefig(
    "outputs/loss_plot.jpg",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ============================================================
# TASK 7
# Evaluate CNN
#
# Accuracy
# Precision
# Recall
# F1 Score
# Confusion Matrix
# Classification Report
# ============================================================

print("\n\n" + "=" * 70)
print("TASK 7 - MODEL EVALUATION")
print("=" * 70)


# ------------------------------------------------------------
# Test set evaluation
# ------------------------------------------------------------

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("\nTest Loss     :", test_loss)
print("Test Accuracy :", test_accuracy)


# ------------------------------------------------------------
# Predictions
# ------------------------------------------------------------

y_probability = model.predict(
    x_test,
    verbose=1
)

y_pred = np.argmax(
    y_probability,
    axis=1
)

y_true = y_test.flatten()


# ------------------------------------------------------------
# Metrics
# ------------------------------------------------------------

accuracy = accuracy_score(
    y_true,
    y_pred
)

precision = precision_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)


print("\nFinal Evaluation Results")
print("-" * 40)

print(
    f"Accuracy  : {accuracy:.4f}"
)

print(
    f"Precision : {precision:.4f}"
)

print(
    f"Recall    : {recall:.4f}"
)

print(
    f"F1-score  : {f1:.4f}"
)

print(
    f"Parameters: {model.count_params()}"
)


# ------------------------------------------------------------
# Classification Report
# ------------------------------------------------------------

print("\nClassification Report:")
print(
    classification_report(
        y_true,
        y_pred,
        target_names=class_names,
        digits=4
    )
)


# ============================================================
# CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_true,
    y_pred
)


plt.figure(figsize=(9, 8))

plt.imshow(
    cm,
    interpolation="nearest",
    cmap="Blues"
)

plt.title("CIFAR-10 Confusion Matrix")

plt.colorbar()

tick_marks = np.arange(len(class_names))

plt.xticks(
    tick_marks,
    class_names,
    rotation=45
)

plt.yticks(
    tick_marks,
    class_names
)

plt.xlabel("Predicted Label")
plt.ylabel("True Label")


# Add numbers inside matrix
threshold = cm.max() / 2

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):

        plt.text(
            j,
            i,
            str(cm[i, j]),
            ha="center",
            va="center",
            color="white" if cm[i, j] > threshold else "black"
        )

plt.tight_layout()

plt.savefig(
    "outputs/confusion_matrix.jpg",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


# ============================================================
# SAVE FINAL RESULTS TO TEXT FILE
# ============================================================

with open(
    "outputs/results.txt",
    "w"
) as f:

    f.write("CS3807 - EXPERIMENT 3 RESULTS\n")
    f.write("=" * 50 + "\n\n")

    f.write("DATASET\n")
    f.write(f"Training samples : {len(x_train)}\n")
    f.write(f"Testing samples  : {len(x_test)}\n")
    f.write(f"Image size       : {x_train.shape[1:]}\n")
    f.write(f"Number of classes: {len(class_names)}\n\n")

    f.write("TASK 2 - KERNEL COMPARISON\n")
    f.write("3x3 -> 30 x 30 x 16\n")
    f.write("5x5 -> 28 x 28 x 16\n")
    f.write("7x7 -> 26 x 26 x 16\n\n")

    f.write("TASK 3 - STRIDE AND PADDING\n")
    f.write("Stride 1, Valid -> 30 x 30 x 16\n")
    f.write("Stride 2, Valid -> 15 x 15 x 16\n")
    f.write("Stride 1, Same  -> 32 x 32 x 16\n")
    f.write("Stride 2, Same  -> 16 x 16 x 16\n\n")

    f.write("TASK 5 - POOLING\n")
    f.write("Max Pooling     -> 16 x 16 x 3\n")
    f.write("Average Pooling -> 16 x 16 x 3\n\n")

    f.write("TASK 7 - FINAL RESULTS\n")
    f.write(f"Test Accuracy : {test_accuracy:.4f}\n")
    f.write(f"Accuracy      : {accuracy:.4f}\n")
    f.write(f"Precision     : {precision:.4f}\n")
    f.write(f"Recall        : {recall:.4f}\n")
    f.write(f"F1-score      : {f1:.4f}\n")
    f.write(f"Parameters    : {model.count_params()}\n")


# ============================================================
# FINAL OUTPUT LIST
# ============================================================

print("\n\n" + "=" * 70)
print("EXPERIMENT COMPLETED")
print("=" * 70)

print("\nAll mandatory JPG plots have been saved in:")
print("outputs/")

print("\nFiles generated:")

files = [
    "sample_images.jpg",
    "class_distribution.jpg",
    "kernel_comparison.jpg",
    "feature_maps.jpg",
    "pooling_comparison.jpg",
    "accuracy_plot.jpg",
    "loss_plot.jpg",
    "confusion_matrix.jpg",
    "results.txt"
]

for file in files:
    print(" -", file)

print("\nFinal Results")
print("-" * 40)
print(f"Training Accuracy : {history.history['accuracy'][-1]:.4f}")
print(f"Validation Accuracy: {history.history['val_accuracy'][-1]:.4f}")
print(f"Testing Accuracy  : {test_accuracy:.4f}")
print(f"Precision         : {precision:.4f}")
print(f"Recall            : {recall:.4f}")
print(f"F1-score          : {f1:.4f}")
print(f"Parameters        : {model.count_params()}")

print("\nDone!")
```
