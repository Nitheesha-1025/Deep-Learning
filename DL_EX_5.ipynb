```python
# ============================================================
# CS3807 - DEEP LEARNING LABORATORY
# EXPERIMENT 5
#
# Comprehensive Study of CNN Training, Regularization,
# Optimization, Hyperparameter Tuning, Transfer Learning
# and Cross-Validation
#
# Dataset: Oxford-IIIT Pet Dataset
# Model: MobileNetV2
# ============================================================


# ============================================================
# 0. IMPORTS
# ============================================================

import os
import time
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import tensorflow as tf

from tensorflow.keras import layers
from tensorflow.keras import models
from tensorflow.keras import regularizers

from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

import seaborn as sns


# ============================================================
# 1. CONFIGURATION
# ============================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)


IMG_SIZE = (224, 224)
NUM_CLASSES = 37

# ------------------------------------------------------------
# For quick testing
# ------------------------------------------------------------

EPOCHS_INITIALIZATION = 5
EPOCHS_REGULARIZATION = 5
EPOCHS_OPTIMIZER = 5
EPOCHS_HYPERPARAMETER = 5
EPOCHS_TRANSFER = 5
EPOCHS_FINETUNE = 5
EPOCHS_FINAL = 10

BATCH_SIZE = 32

# ------------------------------------------------------------
# Set this to None for the complete dataset.
#
# During development you can use:
# MAX_TRAIN_SAMPLES = 3000
#
# For final submission:
# MAX_TRAIN_SAMPLES = None
# ------------------------------------------------------------

MAX_TRAIN_SAMPLES = 3000


# ============================================================
# 2. DOWNLOAD / LOAD OXFORD-IIIT PET DATASET
# ============================================================

print("=" * 70)
print("LOADING OXFORD-IIIT PET DATASET")
print("=" * 70)


dataset_path = tf.keras.utils.get_file(
    "oxford-iiit-pet.tar.gz",
    origin=(
        "https://www.robots.ox.ac.uk/~vgg/data/pets/"
        "data/images.tar.gz"
    ),
    untar=True
)

print("Dataset path:", dataset_path)


# ============================================================
# FIND IMAGE DIRECTORY
# ============================================================

image_dir = dataset_path

if os.path.isfile(image_dir):
    image_dir = os.path.dirname(image_dir)


# Search recursively for jpg files

image_files = []

for root, dirs, files in os.walk(image_dir):

    for file in files:

        if file.lower().endswith(".jpg"):

            image_files.append(
                os.path.join(root, file)
            )


print("Total images found:", len(image_files))


# ============================================================
# CREATE LABELS FROM FILENAMES
# ============================================================

# Oxford-IIIT Pet filenames look like:
#
# Abyssinian_1.jpg
# american_bulldog_1.jpg
# beagle_1.jpg
#
# The part before the final "_number" is the breed.


def get_breed(filename):

    name = os.path.basename(filename)

    name = os.path.splitext(name)[0]

    parts = name.rsplit("_", 1)

    return parts[0]


breeds = sorted(
    list(
        set(
            get_breed(f)
            for f in image_files
        )
    )
)


breed_to_label = {
    breed: i
    for i, breed in enumerate(breeds)
}


labels = np.array([
    breed_to_label[get_breed(f)]
    for f in image_files
])


print("Number of breeds:", len(breeds))


# ============================================================
# LIMIT DATASET FOR DEVELOPMENT
# ============================================================

if MAX_TRAIN_SAMPLES is not None:

    rng = np.random.default_rng(SEED)

    selected_indices = []

    # Stratified-ish selection:
    # select samples from every class

    for class_id in range(NUM_CLASSES):

        class_indices = np.where(
            labels == class_id
        )[0]

        rng.shuffle(class_indices)

        # approximate proportional selection
        selected_indices.extend(
            class_indices[
                :max(
                    1,
                    int(
                        MAX_TRAIN_SAMPLES
                        / NUM_CLASSES
                    )
                )
            ]
        )

    selected_indices = np.array(
        selected_indices
    )

    image_files = [
        image_files[i]
        for i in selected_indices
    ]

    labels = labels[selected_indices]


print("Images used:", len(image_files))


# ============================================================
# 3. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

from sklearn.model_selection import train_test_split


train_files, temp_files, train_labels, temp_labels = \
    train_test_split(
        image_files,
        labels,
        test_size=0.30,
        random_state=SEED,
        stratify=labels
    )


validation_files, test_files, validation_labels, test_labels = \
    train_test_split(
        temp_files,
        temp_labels,
        test_size=0.50,
        random_state=SEED,
        stratify=temp_labels
    )


print("\nDataset split:")
print("Training:", len(train_files))
print("Validation:", len(validation_files))
print("Testing:", len(test_files))


# ============================================================
# IMPORTANT
#
# test_files/test_labels will NOT be used for
# hyperparameter selection.
# ============================================================


# ============================================================
# 4. DATA PIPELINE
# ============================================================

AUTOTUNE = tf.data.AUTOTUNE


def load_image(filename, label):

    image = tf.io.read_file(filename)

    image = tf.image.decode_jpeg(
        image,
        channels=3
    )

    image = tf.image.resize(
        image,
        IMG_SIZE
    )

    image = preprocess_input(
        image
    )

    return image, label


def create_dataset(
    files,
    labels,
    batch_size=32,
    shuffle=False
):

    dataset = tf.data.Dataset.from_tensor_slices(
        (
            files,
            labels
        )
    )

    if shuffle:

        dataset = dataset.shuffle(
            buffer_size=len(files),
            seed=SEED
        )

    dataset = dataset.map(
        load_image,
        num_parallel_calls=AUTOTUNE
    )

    dataset = dataset.batch(
        batch_size
    )

    dataset = dataset.prefetch(
        AUTOTUNE
    )

    return dataset


# ============================================================
# DEFAULT DATASETS
# ============================================================

train_dataset = create_dataset(
    train_files,
    train_labels,
    BATCH_SIZE,
    shuffle=True
)

validation_dataset = create_dataset(
    validation_files,
    validation_labels,
    BATCH_SIZE
)

test_dataset = create_dataset(
    test_files,
    test_labels,
    BATCH_SIZE
)


# ============================================================
# 5. MODEL CREATION FUNCTION
# ============================================================

def create_mobilenet_model(

    initialization="he",

    regularization="none",

    dropout_rate=0.0,

    frozen=True

):

    # --------------------------------------------------------
    # Pretrained MobileNetV2
    # --------------------------------------------------------

    base_model = MobileNetV2(

        weights="imagenet",

        include_top=False,

        input_shape=(
            224,
            224,
            3
        )
    )


    # --------------------------------------------------------
    # Freeze base
    # --------------------------------------------------------

    base_model.trainable = False


    # --------------------------------------------------------
    # Partial unfreezing
    # --------------------------------------------------------

    if frozen == "partial":

        base_model.trainable = True

        # Freeze everything first

        for layer in base_model.layers:

            layer.trainable = False


        # Unfreeze last 30 layers

        for layer in base_model.layers[-30:]:

            layer.trainable = True


    # --------------------------------------------------------
    # Full fine-tuning
    # --------------------------------------------------------

    elif frozen == "full":

        base_model.trainable = True


    # --------------------------------------------------------
    # Regularization
    # --------------------------------------------------------

    if regularization == "l2":

        reg = regularizers.l2(
            0.0001
        )

    else:

        reg = None


    # --------------------------------------------------------
    # Initialization
    # --------------------------------------------------------

    if initialization == "zero":

        initializer = tf.keras.initializers.Zeros()

    elif initialization == "random":

        initializer = tf.keras.initializers.RandomNormal(
            mean=0.0,
            stddev=0.05
        )

    elif initialization == "xavier":

        initializer = tf.keras.initializers.GlorotUniform()

    else:

        initializer = tf.keras.initializers.HeNormal()


    # --------------------------------------------------------
    # Classifier
    # --------------------------------------------------------

    classifier_layers = [

        layers.GlobalAveragePooling2D(),

        layers.Dense(
            128,
            activation="relu",
            kernel_initializer=initializer,
            kernel_regularizer=reg
        )
    ]


    # --------------------------------------------------------
    # Batch Normalization
    # --------------------------------------------------------

    if regularization == "batchnorm":

        classifier_layers.append(
            layers.BatchNormalization()
        )


    # --------------------------------------------------------
    # Dropout
    # --------------------------------------------------------

    if regularization == "dropout":

        classifier_layers.append(

            layers.Dropout(
                dropout_rate
            )
        )


    classifier_layers.append(

        layers.Dense(
            NUM_CLASSES,
            activation="softmax",
            kernel_initializer=initializer
        )
    )


    # --------------------------------------------------------
    # Complete model
    # --------------------------------------------------------

    model = models.Sequential(

        [base_model]
        + classifier_layers
    )


    return model


# ============================================================
# 6. OPTIMIZER FUNCTION
# ============================================================

def get_optimizer(
    name,
    learning_rate=0.001
):

    if name == "SGD":

        return tf.keras.optimizers.SGD(
            learning_rate=learning_rate
        )

    elif name == "Momentum":

        return tf.keras.optimizers.SGD(
            learning_rate=learning_rate,
            momentum=0.9
        )

    elif name == "RMSProp":

        return tf.keras.optimizers.RMSprop(
            learning_rate=learning_rate
        )

    else:

        return tf.keras.optimizers.Adam(
            learning_rate=learning_rate
        )


# ============================================================
# 7. TRAIN MODEL FUNCTION
# ============================================================

def train_model(

    model,

    optimizer_name="Adam",

    learning_rate=0.001,

    epochs=5,

    train_ds=train_dataset,

    val_ds=validation_dataset

):

    optimizer = get_optimizer(
        optimizer_name,
        learning_rate
    )


    model.compile(

        optimizer=optimizer,

        loss="sparse_categorical_crossentropy",

        metrics=["accuracy"]
    )


    start_time = time.time()


    history = model.fit(

        train_ds,

        validation_data=val_ds,

        epochs=epochs,

        verbose=1
    )


    training_time = (
        time.time()
        - start_time
    )


    return history, training_time


# ============================================================
# 8. PLOT UTILITY
# ============================================================

def plot_histories(

    histories,
    metric,
    title,
    ylabel

):

    plt.figure(
        figsize=(10, 6)
    )


    for name, history in histories.items():

        plt.plot(

            history.history[metric],

            label=name
        )


    plt.xlabel("Epoch")

    plt.ylabel(ylabel)

    plt.title(title)

    plt.legend()

    plt.grid()

    plt.show()


# ============================================================
# TASK 1
# WEIGHT INITIALIZATION
#
# Zero
# Random
# Xavier
# He
# ============================================================

print("\n")
print("=" * 70)
print("TASK 1: WEIGHT INITIALIZATION")
print("=" * 70)


initialization_methods = [

    "zero",
    "random",
    "xavier",
    "he"
]


initialization_histories = {}

initialization_results = []


for initialization in initialization_methods:

    print(
        "\nInitialization:",
        initialization
    )


    model = create_mobilenet_model(

        initialization=initialization,

        regularization="none",

        frozen=True
    )


    history, training_time = train_model(

        model,

        optimizer_name="Adam",

        learning_rate=0.001,

        epochs=EPOCHS_INITIALIZATION
    )


    initialization_histories[
        initialization
    ] = history


    best_val_accuracy = max(
        history.history[
            "val_accuracy"
        ]
    )


    final_loss = \
        history.history["loss"][-1]


    initialization_results.append(

        [
            initialization,
            final_loss,
            best_val_accuracy,
            training_time
        ]
    )


    del model

    tf.keras.backend.clear_session()


# ------------------------------------------------------------
# Plot 1
# Training Loss vs Epoch
# ------------------------------------------------------------

plot_histories(

    initialization_histories,

    "loss",

    "Weight Initialization - Training Loss",

    "Training Loss"
)


# ------------------------------------------------------------
# Plot 2
# Validation Accuracy vs Epoch
# ------------------------------------------------------------

plot_histories(

    initialization_histories,

    "val_accuracy",

    "Weight Initialization - Validation Accuracy",

    "Validation Accuracy"
)


initialization_df = pd.DataFrame(

    initialization_results,

    columns=[
        "Initialization",
        "Final Loss",
        "Best Validation Accuracy",
        "Training Time"
    ]
)


print(
    initialization_df.to_string(
        index=False
    )
)


# ============================================================
# TASK 2
# REGULARIZATION
#
# No regularization
# L2
# Dropout
# Batch Normalization
# ============================================================

print("\n")
print("=" * 70)
print("TASK 2: REGULARIZATION")
print("=" * 70)


regularization_configs = {

    "No Regularization": {
        "regularization": "none",
        "dropout_rate": 0.0
    },

    "L2": {
        "regularization": "l2",
        "dropout_rate": 0.0
    },

    "Dropout": {
        "regularization": "dropout",
        "dropout_rate": 0.5
    },

    "Batch Normalization": {
        "regularization": "batchnorm",
        "dropout_rate": 0.0
    }
}


regularization_histories = {}

regularization_results = []


for name, config in regularization_configs.items():

    print(
        "\nRegularization:",
        name
    )


    model = create_mobilenet_model(

        regularization=config[
            "regularization"
        ],

        dropout_rate=config[
            "dropout_rate"
        ],

        frozen=True
    )


    history, training_time = train_model(

        model,

        optimizer_name="Adam",

        learning_rate=0.001,

        epochs=EPOCHS_REGULARIZATION
    )


    regularization_histories[
        name
    ] = history


    best_val_accuracy = max(
        history.history[
            "val_accuracy"
        ]
    )


    regularization_results.append(

        [
            name,
            best_val_accuracy,
            training_time
        ]
    )


    del model

    tf.keras.backend.clear_session()


# ------------------------------------------------------------
# Plot 3
# Training + Validation Accuracy
# ------------------------------------------------------------

for name, history in regularization_histories.items():

    plt.figure(
        figsize=(8, 5)
    )

    plt.plot(
        history.history["accuracy"],
        label="Training Accuracy"
    )

    plt.plot(
        history.history["val_accuracy"],
        label="Validation Accuracy"
    )

    plt.xlabel("Epoch")

    plt.ylabel("Accuracy")

    plt.title(
        f"Regularization: {name}"
    )

    plt.legend()

    plt.grid()

    plt.show()


# ------------------------------------------------------------
# Plot 4
# Training + Validation Loss
# ------------------------------------------------------------

for name, history in regularization_histories.items():

    plt.figure(
        figsize=(8, 5)
    )

    plt.plot(
        history.history["loss"],
        label="Training Loss"
    )

    plt.plot(
        history.history["val_loss"],
        label="Validation Loss"
    )

    plt.xlabel("Epoch")

    plt.ylabel("Loss")

    plt.title(
        f"Regularization: {name}"
    )

    plt.legend()

    plt.grid()

    plt.show()


regularization_df = pd.DataFrame(

    regularization_results,

    columns=[
        "Regularization",
        "Best Validation Accuracy",
        "Training Time"
    ]
)


print(
    regularization_df.to_string(
        index=False
    )
)


# ============================================================
# TASK 3
# BATCH NORMALIZATION
#
# WITH vs WITHOUT BN
# ============================================================

print("\n")
print("=" * 70)
print("TASK 3: BATCH NORMALIZATION")
print("=" * 70)


bn_histories = {}


# ------------------------------------------------------------
# Without BN
# ------------------------------------------------------------

model_without_bn = create_mobilenet_model(

    regularization="none",

    frozen=True
)


history_without_bn, _ = train_model(

    model_without_bn,

    optimizer_name="Adam",

    learning_rate=0.001,

    epochs=EPOCHS_REGULARIZATION
)


bn_histories[
    "Without BN"
] = history_without_bn


del model_without_bn

tf.keras.backend.clear_session()


# ------------------------------------------------------------
# With BN
# ------------------------------------------------------------

model_with_bn = create_mobilenet_model(

    regularization="batchnorm",

    frozen=True
)


history_with_bn, _ = train_model(

    model_with_bn,

    optimizer_name="Adam",

    learning_rate=0.001,

    epochs=EPOCHS_REGULARIZATION
)


bn_histories[
    "With BN"
] = history_with_bn


# ------------------------------------------------------------
# Plot 5
# ------------------------------------------------------------

plot_histories(

    bn_histories,

    "val_accuracy",

    "With vs Without Batch Normalization",

    "Validation Accuracy"
)


# ============================================================
# TASK 4
# OPTIMIZER COMPARISON
#
# SGD
# Momentum
# RMSProp
# Adam
# ============================================================

print("\n")
print("=" * 70)
print("TASK 4: OPTIMIZER COMPARISON")
print("=" * 70)


optimizers = [

    "SGD",
    "Momentum",
    "RMSProp",
    "Adam"
]


optimizer_histories = {}

optimizer_results = []


for optimizer_name in optimizers:

    print(
        "\nOptimizer:",
        optimizer_name
    )


    model = create_mobilenet_model(

        regularization="none",

        frozen=True
    )


    history, training_time = train_model(

        model,

        optimizer_name=optimizer_name,

        learning_rate=0.001,

        epochs=EPOCHS_OPTIMIZER
    )


    optimizer_histories[
        optimizer_name
    ] = history


    best_val_accuracy = max(

        history.history[
            "val_accuracy"
        ]
    )


    final_loss = \
        history.history["loss"][-1]


    optimizer_results.append(

        [
            optimizer_name,
            final_loss,
            best_val_accuracy,
            training_time
        ]
    )


    del model

    tf.keras.backend.clear_session()


# ------------------------------------------------------------
# Plot 6
# Training Loss
# ------------------------------------------------------------

plot_histories(

    optimizer_histories,

    "loss",

    "Training Loss for Different Optimizers",

    "Training Loss"
)


# ------------------------------------------------------------
# Plot 7
# Validation Accuracy
# ------------------------------------------------------------

plot_histories(

    optimizer_histories,

    "val_accuracy",

    "Validation Accuracy for Different Optimizers",

    "Validation Accuracy"
)


optimizer_df = pd.DataFrame(

    optimizer_results,

    columns=[
        "Optimizer",
        "Final Loss",
        "Best Validation Accuracy",
        "Training Time"
    ]
)


print(
    optimizer_df.to_string(
        index=False
    )
)


# ============================================================
# TASK 5
# HYPERPARAMETER TUNING
#
# Learning Rate
# Batch Size
# Dropout
# Optimizer
# ============================================================

print("\n")
print("=" * 70)
print("TASK 5: HYPERPARAMETER TUNING")
print("=" * 70)


# ============================================================
# 5A. LEARNING RATE
# ============================================================

learning_rates = [

    0.001,
    0.0001
]


lr_results = []


for lr in learning_rates:

    print(
        "\nLearning Rate:",
        lr
    )


    model = create_mobilenet_model(

        regularization="none",

        frozen=True
    )


    history, training_time = train_model(

        model,

        optimizer_name="Adam",

        learning_rate=lr,

        epochs=EPOCHS_HYPERPARAMETER
    )


    best_val_accuracy = max(

        history.history[
            "val_accuracy"
        ]
    )


    lr_results.append(

        [
            lr,
            best_val_accuracy,
            training_time
        ]
    )


    del model

    tf.keras.backend.clear_session()


lr_df = pd.DataFrame(

    lr_results,

    columns=[
        "Learning Rate",
        "Validation Accuracy",
        "Training Time"
    ]
)


print(
    lr_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# Plot 8
# ------------------------------------------------------------

plt.figure(
    figsize=(8, 5)
)

plt.plot(

    lr_df["Learning Rate"].astype(str),

    lr_df["Validation Accuracy"],

    marker="o"
)

plt.xlabel("Learning Rate")

plt.ylabel("Validation Accuracy")

plt.title(
    "Learning Rate vs Validation Accuracy"
)

plt.grid()

plt.show()


# ============================================================
# 5B. BATCH SIZE
# ============================================================

batch_sizes = [

    16,
    32,
    64
]


batch_results = []


for batch_size in batch_sizes:

    print(
        "\nBatch Size:",
        batch_size
    )


    batch_train = create_dataset(

        train_files,

        train_labels,

        batch_size,

        shuffle=True
    )


    batch_validation = create_dataset(

        validation_files,

        validation_labels,

        batch_size
    )


    model = create_mobilenet_model(

        regularization="none",

        frozen=True
    )


    history, training_time = train_model(

        model,

        optimizer_name="Adam",

        learning_rate=0.001,

        epochs=EPOCHS_HYPERPARAMETER,

        train_ds=batch_train,

        val_ds=batch_validation
    )


    best_val_accuracy = max(

        history.history[
            "val_accuracy"
        ]
    )


    batch_results.append(

        [
            batch_size,
            best_val_accuracy,
            training_time
        ]
    )


    del model

    tf.keras.backend.clear_session()


batch_df = pd.DataFrame(

    batch_results,

    columns=[
        "Batch Size",
        "Validation Accuracy",
        "Training Time"
    ]
)


print(
    batch_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# Plot 9
# ------------------------------------------------------------

plt.figure(
    figsize=(8, 5)
)

plt.plot(

    batch_df["Batch Size"],

    batch_df["Validation Accuracy"],

    marker="o"
)

plt.xlabel("Batch Size")

plt.ylabel("Validation Accuracy")

plt.title(
    "Batch Size vs Validation Accuracy"
)

plt.grid()

plt.show()


# ============================================================
# 5C. DROPOUT RATE
# ============================================================

dropout_rates = [

    0.0,
    0.25,
    0.5
]


dropout_results = []


for dropout in dropout_rates:

    print(
        "\nDropout:",
        dropout
    )


    if dropout == 0:

        regularization = "none"

    else:

        regularization = "dropout"


    model = create_mobilenet_model(

        regularization=regularization,

        dropout_rate=dropout,

        frozen=True
    )


    history, training_time = train_model(

        model,

        optimizer_name="Adam",

        learning_rate=0.001,

        epochs=EPOCHS_HYPERPARAMETER
    )


    best_val_accuracy = max(

        history.history[
            "val_accuracy"
        ]
    )


    dropout_results.append(

        [
            dropout,
            best_val_accuracy,
            training_time
        ]
    )


    del model

    tf.keras.backend.clear_session()


dropout_df = pd.DataFrame(

    dropout_results,

    columns=[
        "Dropout Rate",
        "Validation Accuracy",
        "Training Time"
    ]
)


print(
    dropout_df.to_string(
        index=False
    )
)


# ------------------------------------------------------------
# Plot 10
# ------------------------------------------------------------

plt.figure(
    figsize=(8, 5)
)

plt.plot(

    dropout_df["Dropout Rate"],

    dropout_df["Validation Accuracy"],

    marker="o"
)

plt.xlabel("Dropout Rate")

plt.ylabel("Validation Accuracy")

plt.title(
    "Dropout Rate vs Validation Accuracy"
)

plt.grid()

plt.show()


# ============================================================
# 5D. FROZEN vs PARTIAL UNFREEZING
# ============================================================

freeze_results = []


for strategy in [

    "frozen",
    "partial"

]:

    print(
        "\nFine-Tuning Strategy:",
        strategy
    )


    model = create_mobilenet_model(

        regularization="none",

        frozen=strategy
    )


    history, training_time = train_model(

        model,

        optimizer_name="Adam",

        learning_rate=0.001,

        epochs=EPOCHS_HYPERPARAMETER
    )


    best_val_accuracy = max(

        history.history[
            "val_accuracy"
        ]
    )


    freeze_results.append(

        [
            strategy,
            best_val_accuracy,
            training_time
        ]
    )


    del model

    tf.keras.backend.clear_session()


freeze_df = pd.DataFrame(

    freeze_results,

    columns=[
        "Strategy",
        "Validation Accuracy",
        "Training Time"
    ]
)


print(
    freeze_df.to_string(
        index=False
    )
)


# ============================================================
# TASK 6
# TRANSFER LEARNING
#
# FEATURE EXTRACTION
# ============================================================

print("\n")
print("=" * 70)
print("TASK 6: TRANSFER LEARNING - FEATURE EXTRACTION")
print("=" * 70)


feature_model = create_mobilenet_model(

    regularization="none",

    frozen=True
)


feature_history, feature_time = train_model(

    feature_model,

    optimizer_name="Adam",

    learning_rate=0.001,

    epochs=EPOCHS_TRANSFER
)


feature_best_accuracy = max(

    feature_history.history[
        "val_accuracy"
    ]
)


# ============================================================
# TASK 7
# FINE-TUNING
#
# Smaller learning rate
# ============================================================

print("\n")
print("=" * 70)
print("TASK 7: FINE-TUNING")
print("=" * 70)


# Start from the feature extraction model

fine_model = feature_model


# Unfreeze last layers

fine_model.get_layer(
    index=0
).trainable = True


# Freeze first layers

base_model = fine_model.get_layer(
    index=0
)


for layer in base_model.layers[:-30]:

    layer.trainable = False


for layer in base_model.layers[-30:]:

    layer.trainable = True


# Recompile with smaller learning rate

fine_model.compile(

    optimizer=tf.keras.optimizers.Adam(

        learning_rate=0.00001
    ),

    loss="sparse_categorical_crossentropy",

    metrics=["accuracy"]
)


start_time = time.time()


fine_history = fine_model.fit(

    train_dataset,

    validation_data=validation_dataset,

    epochs=EPOCHS_FINETUNE,

    verbose=1
)


fine_time = (
    time.time()
    - start_time
)


fine_best_accuracy = max(

    fine_history.history[
        "val_accuracy"
    ]
)


# ============================================================
# PLOT 11
# FEATURE EXTRACTION vs FINE-TUNING
# ============================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(

    feature_history.history[
        "val_accuracy"
    ],

    label="Feature Extraction"
)


plt.plot(

    fine_history.history[
        "val_accuracy"
    ],

    label="Fine-Tuning"
)


plt.xlabel("Epoch")

plt.ylabel("Validation Accuracy")

plt.title(
    "Feature Extraction vs Fine-Tuning"
)

plt.legend()

plt.grid()

plt.show()


# ============================================================
# PLOT 12
# TRAINING AND VALIDATION LOSS
# ============================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(

    fine_history.history["loss"],

    label="Training Loss"
)


plt.plot(

    fine_history.history["val_loss"],

    label="Validation Loss"
)


plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title(
    "Fine-Tuning Training and Validation Loss"
)

plt.legend()

plt.grid()

plt.show()


# ============================================================
# TASK 8
# 5-FOLD CROSS-VALIDATION
# ============================================================

print("\n")
print("=" * 70)
print("TASK 8: 5-FOLD CROSS-VALIDATION")
print("=" * 70)


# ------------------------------------------------------------
# Select promising configurations
# ------------------------------------------------------------

cv_configs = {

    "C1_Baseline": {

        "learning_rate": 0.001,

        "dropout": 0.0,

        "batch_size": 32,

        "optimizer": "Adam",

        "frozen": "frozen"
    },


    "C2_Dropout": {

        "learning_rate": 0.001,

        "dropout": 0.5,

        "batch_size": 32,

        "optimizer": "Adam",

        "frozen": "frozen"
    },


    "C3_LowLR": {

        "learning_rate": 0.0001,

        "dropout": 0.0,

        "batch_size": 32,

        "optimizer": "Adam",

        "frozen": "frozen"
    },


    "C4_FineTune": {

        "learning_rate": 0.00001,

        "dropout": 0.0,

        "batch_size": 32,

        "optimizer": "Adam",

        "frozen": "partial"
    }
}


# ============================================================
# CROSS VALIDATION FUNCTION
# ============================================================

def run_cross_validation(

    config_name,

    config,

    n_splits=5

):

    print("\n")
    print(
        "Cross Validation:",
        config_name
    )


    skf = StratifiedKFold(

        n_splits=n_splits,

        shuffle=True,

        random_state=SEED
    )


    fold_accuracies = []

    fold_times = []


    all_files = np.array(
        train_files
    )

    all_labels = np.array(
        train_labels
    )


    for fold, (train_idx, val_idx) in enumerate(

        skf.split(
            all_files,
            all_labels
        ),

        start=1

    ):

        print(
            "\nFold",
            fold
        )


        fold_train_files = \
            all_files[train_idx]

        fold_train_labels = \
            all_labels[train_idx]


        fold_val_files = \
            all_files[val_idx]

        fold_val_labels = \
            all_labels[val_idx]


        fold_train_ds = create_dataset(

            fold_train_files,

            fold_train_labels,

            config["batch_size"],

            shuffle=True
        )


        fold_val_ds = create_dataset(

            fold_val_files,

            fold_val_labels,

            config["batch_size"]
        )


        if config["dropout"] == 0:

            regularization = "none"

        else:

            regularization = "dropout"


        model = create_mobilenet_model(

            regularization=regularization,

            dropout_rate=config["dropout"],

            frozen=config["frozen"]
        )


        history, training_time = train_model(

            model,

            optimizer_name=config[
                "optimizer"
            ],

            learning_rate=config[
                "learning_rate"
            ],

            epochs=3,

            train_ds=fold_train_ds,

            val_ds=fold_val_ds
        )


        accuracy = max(

            history.history[
                "val_accuracy"
            ]
        )


        fold_accuracies.append(
            accuracy
        )


        fold_times.append(
            training_time
        )


        print(
            "Fold Accuracy:",
            accuracy
        )


        del model

        tf.keras.backend.clear_session()


    mean_accuracy = np.mean(
        fold_accuracies
    )


    std_accuracy = np.std(
        fold_accuracies
    )


    mean_time = np.mean(
        fold_times
    )


    print(
        "\n",
        config_name
    )


    print(
        "Fold Accuracies:",
        fold_accuracies
    )


    print(
        "Mean Accuracy:",
        mean_accuracy
    )


    print(
        "Standard Deviation:",
        std_accuracy
    )


    return (

        fold_accuracies,

        mean_accuracy,

        std_accuracy,

        mean_time
    )


# ============================================================
# RUN 5-FOLD CV
# ============================================================

cv_results = []


for config_name, config in cv_configs.items():

    folds, mean_acc, std_acc, mean_time = (

        run_cross_validation(

            config_name,

            config,

            n_splits=5
        )
    )

    cv_results.append(

        [

            config_name,

            folds[0],

            folds[1],

            folds[2],

            folds[3],

            folds[4],

            mean_acc,

            std_acc,

            mean_time
        ]
    )


cv_df = pd.DataFrame(

    cv_results,

    columns=[

        "Configuration",

        "Fold 1",

        "Fold 2",

        "Fold 3",

        "Fold 4",

        "Fold 5",

        "Mean Accuracy",

        "Standard Deviation",

        "Training Time"
    ]
)


print("\n")
print("=" * 70)
print("5-FOLD CROSS-VALIDATION RESULTS")
print("=" * 70)


print(
    cv_df.to_string(
        index=False
    )
)


# ============================================================
# PLOT 13
# CV ACCURACY WITH STANDARD DEVIATION
# ============================================================

plt.figure(
    figsize=(10, 6)
)


plt.bar(

    cv_df["Configuration"],

    cv_df["Mean Accuracy"],

    yerr=cv_df[
        "Standard Deviation"
    ],

    capsize=5
)


plt.xlabel(
    "Configuration"
)

plt.ylabel(
    "Mean Validation Accuracy"
)

plt.title(
    "5-Fold Cross-Validation Accuracy"
)

plt.xticks(
    rotation=20
)

plt.grid(
    axis="y"
)

plt.show()


# ============================================================
# SELECT BEST CONFIGURATION
# ============================================================

best_index = cv_df[
    "Mean Accuracy"
].idxmax()


best_configuration = cv_df.loc[
    best_index,
    "Configuration"
]


best_config = cv_configs[
    best_configuration
]


print("\n")
print(
    "BEST CONFIGURATION:",
    best_configuration
)


print(
    best_config
)


# ============================================================
# TASK 9
# ADDITIONAL EXERCISE
#
# TWO NEW CONFIGURATIONS
# ============================================================

print("\n")
print("=" * 70)
print("TASK 9: ADDITIONAL EXERCISE")
print("=" * 70)


additional_configs = {

    "New_C1": {

        "learning_rate": 0.0001,

        "dropout": 0.25,

        "batch_size": 16,

        "optimizer": "Adam",

        "frozen": "frozen"
    },


    "New_C2": {

        "learning_rate": 0.00001,

        "dropout": 0.5,

        "batch_size": 64,

        "optimizer": "SGD",

        "frozen": "partial"
    }
}


additional_results = []


for config_name, config in additional_configs.items():

    folds, mean_acc, std_acc, mean_time = (

        run_cross_validation(

            config_name,

            config,

            n_splits=5
        )
    )


    additional_results.append(

        [

            config_name,

            folds[0],

            folds[1],

            folds[2],

            folds[3],

            folds[4],

            mean_acc,

            std_acc,

            mean_time
        ]
    )


additional_df = pd.DataFrame(

    additional_results,

    columns=[

        "Configuration",

        "Fold 1",

        "Fold 2",

        "Fold 3",

        "Fold 4",

        "Fold 5",

        "Mean Accuracy",

        "Standard Deviation",

        "Training Time"
    ]
)


print(
    additional_df.to_string(
        index=False
    )
)


# ============================================================
# TASK 10
# FINAL MODEL
#
# Retrain selected configuration
# on complete training + validation data
#
# TEST SET USED ONLY HERE
# ============================================================

print("\n")
print("=" * 70)
print("TASK 10: FINAL MODEL TRAINING")
print("=" * 70)


# ============================================================
# COMBINE TRAINING + VALIDATION
# ============================================================

final_train_files = np.concatenate(

    [
        np.array(train_files),

        np.array(validation_files)
    ]
)


final_train_labels = np.concatenate(

    [
        np.array(train_labels),

        np.array(validation_labels)
    ]
)


final_train_dataset = create_dataset(

    final_train_files,

    final_train_labels,

    best_config["batch_size"],

    shuffle=True
)


# ============================================================
# CREATE FINAL MODEL
# ============================================================

if best_config["dropout"] == 0:

    final_regularization = "none"

else:

    final_regularization = "dropout"


final_model = create_mobilenet_model(

    regularization=final_regularization,

    dropout_rate=best_config["dropout"],

    frozen=best_config["frozen"]
)


# ============================================================
# TRAIN FINAL MODEL
# ============================================================

final_history, final_training_time = train_model(

    final_model,

    optimizer_name=best_config["optimizer"],

    learning_rate=best_config["learning_rate"],

    epochs=EPOCHS_FINAL,

    train_ds=final_train_dataset,

    val_ds=validation_dataset
)


# ============================================================
# FINAL TEST EVALUATION
# ============================================================

print("\n")
print("=" * 70)
print("FINAL TEST EVALUATION")
print("=" * 70)


test_loss, test_accuracy = final_model.evaluate(

    test_dataset,

    verbose=1
)


print(
    "Test Loss:",
    test_loss
)


print(
    "Test Accuracy:",
    test_accuracy
)


# ============================================================
# PREDICTIONS
# ============================================================

probabilities = final_model.predict(

    test_dataset,

    verbose=1
)


predictions = np.argmax(

    probabilities,

    axis=1
)


true_labels = np.array(
    test_labels
)


# ============================================================
# PRECISION
# ============================================================

precision = precision_score(

    true_labels,

    predictions,

    average="weighted",

    zero_division=0
)


# ============================================================
# RECALL
# ============================================================

recall = recall_score(

    true_labels,

    predictions,

    average="weighted",

    zero_division=0
)


# ============================================================
# F1
# ============================================================

f1 = f1_score(

    true_labels,

    predictions,

    average="weighted",

    zero_division=0
)


print("\n")
print("Precision:", precision)

print("Recall:", recall)

print("F1 Score:", f1)


# ============================================================
# NUMBER OF PARAMETERS
# ============================================================

number_of_parameters = \
    final_model.count_params()


print(
    "Number of Parameters:",
    number_of_parameters
)


print(
    "Training Time:",
    final_training_time,
    "seconds"
)


# ============================================================
# CLASSIFICATION REPORT
# ============================================================

print("\n")
print("=" * 70)
print("CLASSIFICATION REPORT")
print("=" * 70)


print(

    classification_report(

        true_labels,

        predictions,

        target_names=breeds,

        zero_division=0
    )
)


# ============================================================
# PLOT 14
# CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    true_labels,

    predictions
)


plt.figure(

    figsize=(16, 14)
)


sns.heatmap(

    cm,

    annot=False,

    cmap="Blues",

    xticklabels=breeds,

    yticklabels=breeds
)


plt.xlabel(
    "Predicted Class"
)

plt.ylabel(
    "Actual Class"
)

plt.title(
    "Final Model Confusion Matrix"
)

plt.xticks(
    rotation=90
)

plt.yticks(
    rotation=0
)

plt.tight_layout()

plt.show()


# ============================================================
# OPTIONAL PLOT 15
# MISCLASSIFIED IMAGES
# ============================================================

print("\n")
print("=" * 70)
print("MISCLASSIFIED IMAGES")
print("=" * 70)


misclassified_indices = np.where(

    predictions != true_labels

)[0]


# Display up to 10 examples

num_examples = min(

    10,

    len(misclassified_indices)
)


plt.figure(

    figsize=(15, 10)
)


for i in range(num_examples):

    idx = misclassified_indices[i]


    image = tf.io.read_file(

        test_files[idx]
    )


    image = tf.image.decode_jpeg(

        image,

        channels=3
    )


    image = tf.image.resize(

        image,

        IMG_SIZE
    )


    image = image.numpy().astype(
        "uint8"
    )


    plt.subplot(

        2,

        5,

        i + 1
    )


    plt.imshow(image)


    plt.title(

        "True: "
        + breeds[true_labels[idx]]
        + "\nPred: "
        + breeds[predictions[idx]]
    )


    plt.axis("off")


plt.tight_layout()

plt.show()


# ============================================================
# FINAL RESULTS TABLE
# ============================================================

print("\n")
print("=" * 70)
print("FINAL RESULTS")
print("=" * 70)


final_results = pd.DataFrame({

    "Metric": [

        "Mean CV Accuracy",

        "CV Standard Deviation",

        "Test Accuracy",

        "Precision",

        "Recall",

        "F1 Score",

        "Training Time (seconds)",

        "Number of Parameters"
    ],


    "Value": [

        cv_df.loc[
            best_index,
            "Mean Accuracy"
        ],

        cv_df.loc[
            best_index,
            "Standard Deviation"
        ],

        test_accuracy,

        precision,

        recall,

        f1,

        final_training_time,

        number_of_parameters
    ]

})


print(

    final_results.to_string(
        index=False
    )
)


# ============================================================
# OVERALL RESULTS
# ============================================================

print("\n")
print("=" * 70)
print("OVERALL EXPERIMENT RESULTS")
print("=" * 70)


overall_results = pd.DataFrame({

    "Stage": [

        "Baseline",

        "Best Initialization",

        "Best Regularization",

        "Best Optimizer",

        "Best Hyperparameters",

        "Fine-Tuned Model",

        "Final Model"
    ],


    "Validation/Test Accuracy": [

        max(
            history_without_bn.history[
                "val_accuracy"
            ]
        ),

        initialization_df[
            "Best Validation Accuracy"
        ].max(),

        regularization_df[
            "Best Validation Accuracy"
        ].max(),

        optimizer_df[
            "Best Validation Accuracy"
        ].max(),

        max(
            lr_df[
                "Validation Accuracy"
            ].max(),

            batch_df[
                "Validation Accuracy"
            ].max(),

            dropout_df[
                "Validation Accuracy"
            ].max(),

            freeze_df[
                "Validation Accuracy"
            ].max()
        ),

        fine_best_accuracy,

        test_accuracy
    ]

})


print(

    overall_results.to_string(
        index=False
    )
)


# ============================================================
# SAVE RESULTS
# ============================================================

initialization_df.to_csv(
    "initialization_results.csv",
    index=False
)


regularization_df.to_csv(
    "regularization_results.csv",
    index=False
)


optimizer_df.to_csv(
    "optimizer_results.csv",
    index=False
)


lr_df.to_csv(
    "learning_rate_results.csv",
    index=False
)


batch_df.to_csv(
    "batch_size_results.csv",
    index=False
)


dropout_df.to_csv(
    "dropout_results.csv",
    index=False
)


freeze_df.to_csv(
    "fine_tuning_results.csv",
    index=False
)


cv_df.to_csv(
    "cross_validation_results.csv",
    index=False
)


additional_df.to_csv(
    "additional_exercise_results.csv",
    index=False
)


final_results.to_csv(
    "final_results.csv",
    index=False
)


print("\n")
print("=" * 70)
print("EXPERIMENT COMPLETED")
print("=" * 70)

print(
    "All result tables have been saved as CSV files."
)
```
