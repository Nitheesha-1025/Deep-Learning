```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from tensorflow.keras import layers
from tensorflow.keras import models
from tensorflow.keras.applications import ResNet50

from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report


# ============================================================
# TASK 1: DATASET PREPARATION
# ============================================================

# Load CIFAR-10 dataset

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()


print("Training data:", x_train.shape)
print("Testing data:", x_test.shape)


# Normalize pixel values to [0,1]

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0


# CIFAR-10 class names

class_names = [
    "Airplane",
    "Automobile",
    "Bird",
    "Cat",
    "Deer",
    "Dog",
    "Frog",
    "Horse",
    "Ship",
    "Truck"
]


# Display 10 sample images

plt.figure(figsize=(15, 5))

for i in range(10):

    plt.subplot(2, 5, i + 1)

    plt.imshow(x_train[i])

    plt.title(class_names[y_train[i][0]])

    plt.axis("off")

plt.tight_layout()

plt.show()


# ============================================================
# CONVERT LABELS
# ============================================================

y_train = tf.keras.utils.to_categorical(
    y_train,
    10
)

y_test = tf.keras.utils.to_categorical(
    y_test,
    10
)


# ============================================================
# CREATE DATASETS
# ============================================================

train_dataset = tf.data.Dataset.from_tensor_slices(
    (x_train, y_train)
)

test_dataset = tf.data.Dataset.from_tensor_slices(
    (x_test, y_test)
)


# Resize each image when it is loaded

def resize_image(image, label):

    image = tf.image.resize(
        image,
        (224, 224)
    )

    return image, label


train_dataset = train_dataset.map(
    resize_image,
    num_parallel_calls=tf.data.AUTOTUNE
)

test_dataset = test_dataset.map(
    resize_image,
    num_parallel_calls=tf.data.AUTOTUNE
)


# Batch the data

train_dataset = train_dataset.batch(32).prefetch(
    tf.data.AUTOTUNE
)

test_dataset = test_dataset.batch(32).prefetch(
    tf.data.AUTOTUNE
)


# ============================================================
# TASK 2: TRANSFER LEARNING
# ============================================================

base_model = ResNet50(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)


# Freeze convolutional base

base_model.trainable = False


# Create model

model = models.Sequential([

    base_model,

    layers.GlobalAveragePooling2D(),

    layers.Dense(
        128,
        activation="relu"
    ),

    layers.Dense(
        10,
        activation="softmax"
    )
])


model.summary()


# ============================================================
# TASK 3: MODEL TRAINING
# ============================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="categorical_crossentropy",

    metrics=["accuracy"]
)


history = model.fit(

    train_dataset,

    validation_data=test_dataset,

    epochs=10
)


# ============================================================
# TRAINING ACCURACY
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"]
)

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.title("Training Accuracy")

plt.grid()

plt.show()


# ============================================================
# VALIDATION ACCURACY
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["val_accuracy"]
)

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.title("Validation Accuracy")

plt.grid()

plt.show()


# ============================================================
# TRAINING LOSS
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"]
)

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title("Training Loss")

plt.grid()

plt.show()


# ============================================================
# VALIDATION LOSS
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["val_loss"]
)

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title("Validation Loss")

plt.grid()

plt.show()


# ============================================================
# TASK 5: EVALUATION
# ============================================================

test_loss, test_accuracy = model.evaluate(
    test_dataset
)


print("\n====================================")
print("BEFORE FINE-TUNING")
print("====================================")

print("Test Loss:", test_loss)

print("Test Accuracy:", test_accuracy)


# ============================================================
# PREDICTIONS
# ============================================================

y_probability = model.predict(
    test_dataset
)


y_pred = np.argmax(
    y_probability,
    axis=1
)


y_true = np.argmax(
    y_test,
    axis=1
)


# ============================================================
# PRECISION
# ============================================================

precision = precision_score(
    y_true,
    y_pred,
    average="weighted"
)

print("Precision:", precision)


# ============================================================
# RECALL
# ============================================================

recall = recall_score(
    y_true,
    y_pred,
    average="weighted"
)

print("Recall:", recall)


# ============================================================
# F1 SCORE
# ============================================================

f1 = f1_score(
    y_true,
    y_pred,
    average="weighted"
)

print("F1 Score:", f1)


# ============================================================
# CLASSIFICATION REPORT
# ============================================================

print("\n====================================")
print("CLASSIFICATION REPORT")
print("====================================")

print(
    classification_report(
        y_true,
        y_pred,
        target_names=class_names
    )
)


# ============================================================
# CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_true,
    y_pred
)


plt.figure(figsize=(10, 8))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.xlabel("Predicted Class")

plt.ylabel("Actual Class")

plt.title("Confusion Matrix Before Fine-Tuning")

plt.show()


# ============================================================
# TASK 4: FINE-TUNING
# ============================================================

# Unfreeze ResNet50

base_model.trainable = True


# Freeze all layers first

for layer in base_model.layers:

    layer.trainable = False


# Unfreeze last convolutional block

set_trainable = False

for layer in base_model.layers:

    if layer.name.startswith("conv5"):

        set_trainable = True

    if set_trainable:

        layer.trainable = True


# ============================================================
# COMPILE AGAIN
# ============================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="categorical_crossentropy",

    metrics=["accuracy"]
)


# ============================================================
# FINE-TUNE
# ============================================================

fine_history = model.fit(

    train_dataset,

    validation_data=test_dataset,

    epochs=5
)


# ============================================================
# FINE-TUNING ACCURACY
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    fine_history.history["accuracy"]
)

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.title("Fine-Tuning Training Accuracy")

plt.grid()

plt.show()


# ============================================================
# FINE-TUNING VALIDATION ACCURACY
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    fine_history.history["val_accuracy"]
)

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.title("Fine-Tuning Validation Accuracy")

plt.grid()

plt.show()


# ============================================================
# FINE-TUNING TRAINING LOSS
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    fine_history.history["loss"]
)

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title("Fine-Tuning Training Loss")

plt.grid()

plt.show()


# ============================================================
# FINE-TUNING VALIDATION LOSS
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    fine_history.history["val_loss"]
)

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title("Fine-Tuning Validation Loss")

plt.grid()

plt.show()


# ============================================================
# FINAL EVALUATION
# ============================================================

final_loss, final_accuracy = model.evaluate(
    test_dataset
)


print("\n====================================")
print("AFTER FINE-TUNING")
print("====================================")

print("Final Test Loss:", final_loss)

print("Final Test Accuracy:", final_accuracy)


# ============================================================
# FINAL PREDICTIONS
# ============================================================

final_probability = model.predict(
    test_dataset
)


final_pred = np.argmax(
    final_probability,
    axis=1
)


final_true = np.argmax(
    y_test,
    axis=1
)


# ============================================================
# FINAL METRICS
# ============================================================

final_precision = precision_score(
    final_true,
    final_pred,
    average="weighted"
)


final_recall = recall_score(
    final_true,
    final_pred,
    average="weighted"
)


final_f1 = f1_score(
    final_true,
    final_pred,
    average="weighted"
)


print("Final Precision:", final_precision)

print("Final Recall:", final_recall)

print("Final F1 Score:", final_f1)


# ============================================================
# FINAL CLASSIFICATION REPORT
# ============================================================

print("\n====================================")
print("FINAL CLASSIFICATION REPORT")
print("====================================")

print(
    classification_report(
        final_true,
        final_pred,
        target_names=class_names
    )
)


# ============================================================
# FINAL CONFUSION MATRIX
# ============================================================

final_cm = confusion_matrix(
    final_true,
    final_pred
)


plt.figure(figsize=(10, 8))

sns.heatmap(
    final_cm,
    annot=True,
    fmt="d",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.xlabel("Predicted Class")

plt.ylabel("Actual Class")

plt.title("Confusion Matrix After Fine-Tuning")

plt.show()


# ============================================================
# FINAL COMPARISON
# ============================================================

print("\n====================================")
print("TRANSFER LEARNING VS FINE-TUNING")
print("====================================")

print(
    "Accuracy Before Fine-Tuning:",
    test_accuracy
)

print(
    "Accuracy After Fine-Tuning:",
    final_accuracy
)

print(
    "Accuracy Improvement:",
    final_accuracy - test_accuracy
)
```

```python
# ============================================================
# TASK 6: HYPERPARAMETER STUDY
# ============================================================

import pandas as pd
import tensorflow as tf

from tensorflow.keras import layers
from tensorflow.keras import models
from tensorflow.keras.applications import ResNet50
# ============================================================
# TASK 6: HYPERPARAMETER STUDY
# ============================================================

import tensorflow as tf
import numpy as np
import pandas as pd

from tensorflow.keras import layers
from tensorflow.keras import models
from tensorflow.keras.applications import ResNet50


# ============================================================
# LOAD CIFAR-10 DATASET
# ============================================================

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()


# Normalize images

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0


# Convert labels to categorical

y_train = tf.keras.utils.to_categorical(
    y_train,
    10
)

y_test = tf.keras.utils.to_categorical(
    y_test,
    10
)


print("Training data:", x_train.shape)
print("Testing data:", x_test.shape)


# ============================================================
# CREATE DATASETS
# ============================================================

def resize_image(image, label):

    image = tf.image.resize(
        image,
        (224, 224)
    )

    return image, label


train_data = tf.data.Dataset.from_tensor_slices(
    (x_train, y_train)
)

train_data = train_data.map(
    resize_image,
    num_parallel_calls=tf.data.AUTOTUNE
)


test_data = tf.data.Dataset.from_tensor_slices(
    (x_test, y_test)
)

test_data = test_data.map(
    resize_image,
    num_parallel_calls=tf.data.AUTOTUNE
)


# ============================================================
# CREATE MEMORY-EFFICIENT DATASETS
# ============================================================

def resize_image(image, label):

    image = tf.image.resize(
        image,
        (224, 224)
    )

    return image, label


# Training dataset

train_data = tf.data.Dataset.from_tensor_slices(
    (x_train, y_train)
)

train_data = train_data.map(
    resize_image,
    num_parallel_calls=tf.data.AUTOTUNE
)


# Testing dataset

test_data = tf.data.Dataset.from_tensor_slices(
    (x_test, y_test)
)

test_data = test_data.map(
    resize_image,
    num_parallel_calls=tf.data.AUTOTUNE
)


# ============================================================
# FUNCTION TO CREATE RESNET50 MODEL
# ============================================================

def create_model(dense_unit=128, frozen="All"):

    base = ResNet50(
        weights="imagenet",
        include_top=False,
        input_shape=(224, 224, 3)
    )


    # Freeze all layers

    base.trainable = False


    # Partial freezing

    if frozen == "Partial":

        base.trainable = True

        for layer in base.layers:
            layer.trainable = False

        set_trainable = False

        for layer in base.layers:

            if layer.name.startswith("conv5"):

                set_trainable = True

            if set_trainable:

                layer.trainable = True


    # Create complete model

    model = models.Sequential([

        base,

        layers.GlobalAveragePooling2D(),

        layers.Dense(
            dense_unit,
            activation="relu"
        ),

        layers.Dense(
            10,
            activation="softmax"
        )

    ])

    return model


# ============================================================
# FUNCTION TO SELECT OPTIMIZER
# ============================================================

def get_optimizer(
    optimizer_name,
    learning_rate
):

    if optimizer_name == "Adam":

        return tf.keras.optimizers.Adam(
            learning_rate=learning_rate
        )

    else:

        return tf.keras.optimizers.SGD(
            learning_rate=learning_rate
        )


# ============================================================
# FUNCTION TO RUN ONE EXPERIMENT
# ============================================================

def run_experiment(
    learning_rate=0.001,
    batch_size=32,
    epochs=10,
    optimizer_name="Adam",
    dense_unit=128,
    frozen="All"
):

    print(
        "\nRunning:",
        "LR =", learning_rate,
        "| Batch =", batch_size,
        "| Epochs =", epochs,
        "| Optimizer =", optimizer_name,
        "| Dense =", dense_unit,
        "| Frozen =", frozen
    )


    # Create new model

    model = create_model(
        dense_unit,
        frozen
    )


    # Select optimizer

    optimizer = get_optimizer(
        optimizer_name,
        learning_rate
    )


    # Compile

    model.compile(

        optimizer=optimizer,

        loss="categorical_crossentropy",

        metrics=["accuracy"]

    )


    # Create batches

    train_batch = train_data.batch(
        batch_size
    ).prefetch(
        tf.data.AUTOTUNE
    )


    test_batch = test_data.batch(
        32
    ).prefetch(
        tf.data.AUTOTUNE
    )


    # Train

    model.fit(

        train_batch,

        validation_data=test_batch,

        epochs=epochs,

        verbose=0

    )


    # Evaluate

    loss, accuracy = model.evaluate(

        test_batch,

        verbose=0

    )


    print(
        "Accuracy:",
        accuracy
    )


    # Delete model to free GPU memory

    del model

    tf.keras.backend.clear_session()


    return accuracy


# ============================================================
# HYPERPARAMETERS
# ============================================================

learning_rates = [
    0.001,
    0.0001
]

batch_sizes = [
    16,
    32,
    64
]

epoch_values = [
    10,
    20
]

optimizers = [
    "Adam",
    "SGD"
]

dense_units = [
    128,
    256
]

frozen_layers = [
    "All",
    "Partial"
]


# ============================================================
# STORE RESULTS
# ============================================================

results = []


# ============================================================
# 1. LEARNING RATE
# ============================================================

for lr in learning_rates:

    accuracy = run_experiment(
        learning_rate=lr
    )

    results.append([
        "Learning Rate",
        lr,
        accuracy
    ])


# ============================================================
# 2. BATCH SIZE
# ============================================================

for batch in batch_sizes:

    accuracy = run_experiment(
        batch_size=batch
    )

    results.append([
        "Batch Size",
        batch,
        accuracy
    ])


# ============================================================
# 3. EPOCHS
# ============================================================

for epoch in epoch_values:

    accuracy = run_experiment(
        epochs=epoch
    )

    results.append([
        "Epochs",
        epoch,
        accuracy
    ])


# ============================================================
# 4. OPTIMIZER
# ============================================================

for optimizer in optimizers:

    accuracy = run_experiment(
        optimizer_name=optimizer
    )

    results.append([
        "Optimizer",
        optimizer,
        accuracy
    ])


# ============================================================
# 5. DENSE UNITS
# ============================================================

for units in dense_units:

    accuracy = run_experiment(
        dense_unit=units
    )

    results.append([
        "Dense Units",
        units,
        accuracy
    ])


# ============================================================
# 6. FROZEN LAYERS
# ============================================================

for frozen in frozen_layers:

    accuracy = run_experiment(
        frozen=frozen
    )

    results.append([
        "Frozen Layers",
        frozen,
        accuracy
    ])


# ============================================================
# FINAL RESULTS TABLE
# ============================================================

results_df = pd.DataFrame(

    results,

    columns=[
        "Hyperparameter",
        "Setting",
        "Accuracy"
    ]

)


print("\n========================================")
print("HYPERPARAMETER STUDY RESULTS")
print("========================================")

print(
    results_df.to_string(
        index=False
    )
)


# ============================================================
# BEST VALUE FOR EACH HYPERPARAMETER
# ============================================================

print("\n========================================")
print("BEST SETTING")
print("========================================")


for parameter in results_df["Hyperparameter"].unique():

    temp = results_df[
        results_df["Hyperparameter"] == parameter
    ]

    best = temp.loc[
        temp["Accuracy"].idxmax()
    ]

    print(
        parameter,
        "->",
        best["Setting"],
        "Accuracy:",
        best["Accuracy"]
    )
```

```python
# ============================================================
# ADDITIONAL EXERCISES
# 1. VGG16 Transfer Learning
# 3. Adam vs SGD Comparison
# 6. LeNet vs AlexNet vs ResNet50 Comparison
# ============================================================

import tensorflow as tf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from tensorflow.keras import layers
from tensorflow.keras import models
from tensorflow.keras.applications import VGG16
from tensorflow.keras.applications import ResNet50


# ============================================================
# 1. LOAD CIFAR-10 DATASET
# ============================================================

print("Loading CIFAR-10 dataset...")

(x_train, y_train), (x_test, y_test) = \
    tf.keras.datasets.cifar10.load_data()

# Normalize
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Convert labels to categorical
y_train = tf.keras.utils.to_categorical(y_train, 10)
y_test = tf.keras.utils.to_categorical(y_test, 10)

print("Training data:", x_train.shape)
print("Testing data:", x_test.shape)


# ============================================================
# 2. CREATE DATASETS FOR TRANSFER LEARNING
# ============================================================

def resize_image(image, label):

    image = tf.image.resize(
        image,
        (224, 224)
    )

    return image, label


train_transfer = tf.data.Dataset.from_tensor_slices(
    (x_train, y_train)
)

train_transfer = train_transfer.map(
    resize_image,
    num_parallel_calls=tf.data.AUTOTUNE
)

train_transfer = train_transfer.batch(32).prefetch(
    tf.data.AUTOTUNE
)


test_transfer = tf.data.Dataset.from_tensor_slices(
    (x_test, y_test)
)

test_transfer = test_transfer.map(
    resize_image,
    num_parallel_calls=tf.data.AUTOTUNE
)

test_transfer = test_transfer.batch(32).prefetch(
    tf.data.AUTOTUNE
)


# ============================================================
# EXERCISE 1
# TRANSFER LEARNING USING VGG16
# ============================================================

print("\n")
print("=" * 60)
print("EXERCISE 1: VGG16 TRANSFER LEARNING")
print("=" * 60)


# Load pretrained VGG16
vgg_base = VGG16(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)


# Freeze convolutional layers
vgg_base.trainable = False


# Create VGG16 model
vgg_model = models.Sequential([

    vgg_base,

    layers.GlobalAveragePooling2D(),

    layers.Dense(
        128,
        activation="relu"
    ),

    layers.Dense(
        10,
        activation="softmax"
    )
])


# Compile
vgg_model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="categorical_crossentropy",

    metrics=["accuracy"]
)


# Display model
vgg_model.summary()


# Train VGG16
print("\nTraining VGG16...")

vgg_history = vgg_model.fit(

    train_transfer,

    validation_data=test_transfer,

    epochs=5,

    verbose=1
)


# Evaluate VGG16
vgg_loss, vgg_accuracy = vgg_model.evaluate(
    test_transfer,
    verbose=0
)


print("\nVGG16 Results")
print("-----------------------------")
print("Test Loss:", vgg_loss)
print("Test Accuracy:", vgg_accuracy)


# ============================================================
# VGG16 TRAINING CURVES
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    vgg_history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    vgg_history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("VGG16 Training and Validation Accuracy")
plt.legend()
plt.grid()
plt.show()


# ============================================================
# EXERCISE 3
# ADAM VS SGD
# ============================================================

print("\n")
print("=" * 60)
print("EXERCISE 3: ADAM VS SGD")
print("=" * 60)


# ------------------------------------------------------------
# Function to create ResNet50 model
# ------------------------------------------------------------

def create_resnet_model():

    base = ResNet50(

        weights="imagenet",

        include_top=False,

        input_shape=(224, 224, 3)
    )

    # Freeze pretrained layers
    base.trainable = False

    model = models.Sequential([

        base,

        layers.GlobalAveragePooling2D(),

        layers.Dense(
            128,
            activation="relu"
        ),

        layers.Dense(
            10,
            activation="softmax"
        )
    ])

    return model


# ------------------------------------------------------------
# Function to train with selected optimizer
# ------------------------------------------------------------

def train_optimizer_experiment(optimizer_name):

    print("\nTraining with:", optimizer_name)

    model = create_resnet_model()


    # Select optimizer

    if optimizer_name == "Adam":

        optimizer = tf.keras.optimizers.Adam(
            learning_rate=0.001
        )

    else:

        optimizer = tf.keras.optimizers.SGD(
            learning_rate=0.001
        )


    # Compile

    model.compile(

        optimizer=optimizer,

        loss="categorical_crossentropy",

        metrics=["accuracy"]
    )


    # Train

    history = model.fit(

        train_transfer,

        validation_data=test_transfer,

        epochs=5,

        verbose=1
    )


    # Evaluate

    loss, accuracy = model.evaluate(

        test_transfer,

        verbose=0
    )


    print(
        optimizer_name,
        "Accuracy:",
        accuracy
    )


    return model, history, accuracy


# ============================================================
# TRAIN WITH ADAM
# ============================================================

adam_model, adam_history, adam_accuracy = \
    train_optimizer_experiment("Adam")


# ============================================================
# CLEAR SESSION BEFORE SGD
# ============================================================

tf.keras.backend.clear_session()


# ============================================================
# TRAIN WITH SGD
# ============================================================

sgd_model, sgd_history, sgd_accuracy = \
    train_optimizer_experiment("SGD")


# ============================================================
# ADAM VS SGD RESULTS
# ============================================================

optimizer_results = pd.DataFrame({

    "Optimizer": [
        "Adam",
        "SGD"
    ],

    "Test Accuracy": [
        adam_accuracy,
        sgd_accuracy
    ]
})


print("\n")
print("=" * 60)
print("ADAM VS SGD RESULTS")
print("=" * 60)

print(
    optimizer_results.to_string(
        index=False
    )
)


# ============================================================
# ADAM VS SGD GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    adam_history.history["val_accuracy"],
    label="Adam"
)

plt.plot(
    sgd_history.history["val_accuracy"],
    label="SGD"
)

plt.xlabel("Epoch")
plt.ylabel("Validation Accuracy")
plt.title("Adam vs SGD")
plt.legend()
plt.grid()
plt.show()


# ============================================================
# EXERCISE 6
# LENET VS ALEXNET VS RESNET50
# ============================================================

print("\n")
print("=" * 60)
print("EXERCISE 6: LENET VS ALEXNET VS RESNET50")
print("=" * 60)


# ============================================================
# LENET
# ============================================================

def create_lenet():

    model = models.Sequential([

        layers.Conv2D(
            6,
            kernel_size=(5, 5),
            activation="relu",
            input_shape=(32, 32, 3)
        ),

        layers.AveragePooling2D(
            pool_size=(2, 2)
        ),

        layers.Conv2D(
            16,
            kernel_size=(5, 5),
            activation="relu"
        ),

        layers.AveragePooling2D(
            pool_size=(2, 2)
        ),

        layers.Flatten(),

        layers.Dense(
            120,
            activation="relu"
        ),

        layers.Dense(
            84,
            activation="relu"
        ),

        layers.Dense(
            10,
            activation="softmax"
        )
    ])

    return model


# ============================================================
# ALEXNET
# ============================================================

def create_alexnet():

    model = models.Sequential([

        layers.Conv2D(
            96,
            kernel_size=(3, 3),
            padding="same",
            activation="relu",
            input_shape=(32, 32, 3)
        ),

        layers.MaxPooling2D(
            pool_size=(2, 2)
        ),

        layers.Conv2D(
            256,
            kernel_size=(3, 3),
            padding="same",
            activation="relu"
        ),

        layers.MaxPooling2D(
            pool_size=(2, 2)
        ),

        layers.Conv2D(
            384,
            kernel_size=(3, 3),
            padding="same",
            activation="relu"
        ),

        layers.Conv2D(
            384,
            kernel_size=(3, 3),
            padding="same",
            activation="relu"
        ),

        layers.Conv2D(
            256,
            kernel_size=(3, 3),
            padding="same",
            activation="relu"
        ),

        layers.MaxPooling2D(
            pool_size=(2, 2)
        ),

        layers.Flatten(),

        layers.Dense(
            4096,
            activation="relu"
        ),

        layers.Dropout(0.5),

        layers.Dense(
            4096,
            activation="relu"
        ),

        layers.Dropout(0.5),

        layers.Dense(
            10,
            activation="softmax"
        )
    ])

    return model


# ============================================================
# TRAIN LENET
# ============================================================

print("\nTraining LeNet...")

lenet_model = create_lenet()

lenet_model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="categorical_crossentropy",

    metrics=["accuracy"]
)

lenet_history = lenet_model.fit(

    x_train,
    y_train,

    validation_data=(
        x_test,
        y_test
    ),

    epochs=10,

    batch_size=64,

    verbose=1
)

lenet_loss, lenet_accuracy = lenet_model.evaluate(

    x_test,
    y_test,

    verbose=0
)

print(
    "LeNet Accuracy:",
    lenet_accuracy
)


# ============================================================
# CLEAR MEMORY
# ============================================================

tf.keras.backend.clear_session()


# ============================================================
# TRAIN ALEXNET
# ============================================================

print("\nTraining AlexNet...")

alexnet_model = create_alexnet()

alexnet_model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="categorical_crossentropy",

    metrics=["accuracy"]
)

alexnet_history = alexnet_model.fit(

    x_train,
    y_train,

    validation_data=(
        x_test,
        y_test
    ),

    epochs=10,

    batch_size=64,

    verbose=1
)

alexnet_loss, alexnet_accuracy = \
    alexnet_model.evaluate(

        x_test,
        y_test,

        verbose=0
    )

print(
    "AlexNet Accuracy:",
    alexnet_accuracy
)


# ============================================================
# RESNET50 FOR COMPARISON
# ============================================================

print("\nTraining ResNet50...")

resnet_model = create_resnet_model()

resnet_model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),

    loss="categorical_crossentropy",

    metrics=["accuracy"]
)


resnet_history = resnet_model.fit(

    train_transfer,

    validation_data=test_transfer,

    epochs=5,

    verbose=1
)


resnet_loss, resnet_accuracy = \
    resnet_model.evaluate(

        test_transfer,

        verbose=0
    )


print(
    "ResNet50 Accuracy:",
    resnet_accuracy
)


# ============================================================
# FINAL MODEL COMPARISON
# ============================================================

comparison_results = pd.DataFrame({

    "Model": [
        "LeNet",
        "AlexNet",
        "ResNet50"
    ],

    "Test Accuracy": [
        lenet_accuracy,
        alexnet_accuracy,
        resnet_accuracy
    ],

    "Test Loss": [
        lenet_loss,
        alexnet_loss,
        resnet_loss
    ]
})


print("\n")
print("=" * 60)
print("LENET VS ALEXNET VS RESNET50")
print("=" * 60)

print(
    comparison_results.to_string(
        index=False
    )
)


# ============================================================
# MODEL COMPARISON GRAPH
# ============================================================

plt.figure(figsize=(8, 5))

plt.bar(

    comparison_results["Model"],

    comparison_results["Test Accuracy"]
)

plt.xlabel("Model")
plt.ylabel("Test Accuracy")
plt.title("LeNet vs AlexNet vs ResNet50")
plt.ylim(0, 1)
plt.grid(axis="y")

plt.show()


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n")
print("=" * 60)
print("ADDITIONAL EXERCISES SUMMARY")
print("=" * 60)

print(
    "VGG16 Accuracy:",
    vgg_accuracy
)

print(
    "Adam Accuracy:",
    adam_accuracy
)

print(
    "SGD Accuracy:",
    sgd_accuracy
)

print(
    "LeNet Accuracy:",
    lenet_accuracy
)

print(
    "AlexNet Accuracy:",
    alexnet_accuracy
)

print(
    "ResNet50 Accuracy:",
    resnet_accuracy
)
```
